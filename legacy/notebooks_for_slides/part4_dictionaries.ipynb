````python
# Run this cell to see dictionary keys and values in action

car = {"brand": "Toyota", "model": "Camry", "year": 2020, "color": "blue"}
print("Car dictionary:", car)

# Accessing values by key
print("Brand:", car["brand"])
print("Year:", car["year"])

# Getting all keys, values, and items
print("All keys:", list(car.keys()))
print("All values:", list(car.values()))
print("All items:", list(car.items()))

# Checking if a key exists
print("Has 'model' key:", "model" in car)
print("Has 'price' key:", "price" in car)

# Adding a new key-value pair
car["price"] = 25000
print("After adding price:", car)
````

### Adding key-value pairs to dictionaries

Once we've defined a dictionary, it's easy to **add a key-value pair**. We just use an assignment statement, assigning the value to a new key:

**Example:**
```python
dct = {"a": 1, "b": 2}
dct["c"] = 3  # adds new key-value pair
print(dct)  # outputs: {"a": 1, "b": 2, "c": 3}
```

We can also **modify existing values** in the same way:
```python
dct["a"] = 10  # changes existing value
print(dct)  # outputs: {"a": 10, "b": 2, "c": 3}
```

**Key point:** If the key already exists, we'll update its value. If it doesn't exist, we'll create a new key-value pair.

### Typing long lists and dictionaries

It's hard to read code that has very long lines, because we have to use a scroll bar. Fortunately, there are ways to break lines when creating lists and dictionaries.

**Rule:** We can break anywhere **except before a comma**.

Code can be easier to read when items are stacked as in the second example below, though this is not commonly done for small dictionaries:

````python
inventory1 = {"apples": 50, "bananas": 30, "oranges": 25}

inventory2 = {
    "apples": 50,
    "bananas": 30,
    "oranges": 25
}

print(inventory1 == inventory2)
````

### Lists vs. Dictionaries: When to Use Which?

| Feature | Lists | Dictionaries |
|---------|-------|------------|
| **Structure** | Ordered sequence of elements | Unordered collection of key-value pairs |
| **Access Method** | By index (position): `my_list[0]` | By key (name): `my_dict["key"]` |
| **Indexing** | Numeric (0, 1, 2, ...) | Custom keys (strings, numbers, etc.) |
| **Order** | Maintains insertion order | Order not guaranteed (Python 3.7+ maintains insertion order) |
| **Duplicates** | Allows duplicate values | Keys must be unique, values can duplicate |
| **Best Used For** | - Sequences of similar items<br>- When order matters<br>- Mathematical operations | - Key-value relationships<br>- Looking up by name/ID<br>- Configuration settings |
| **Example Use Cases** | - Shopping list<br>- Test scores<br>- Time series data | - Student records<br>- Phone book<br>- Inventory with product codes |
