### Dictionaries

**Dictionaries** are another fundamental data structure in Python. A **dictionary** is an **unordered** collection of **key-value pairs**. Each key is unique and is used to access its corresponding value.

**Key features of dictionaries:**
* Created using **curly braces** `{}` with key-value pairs separated by commas
* Each key-value pair is written as `key: value`
* Keys must be **unique**, but values can be duplicated
* Access values by their **key name** rather than position

As with other variables, we can name dictionaries pretty much anything we want, but we cannot use `dict` because it is a reserved Python keyword.

### Dictionaries vs lists

In a dictionary, you **access values using their corresponding keys**:

`dict[key]` - gets the value for a specific key

**Remember the key difference:**
* **Lists:** Elements are accessed based on their **position** in the list
* **Dictionaries:** Elements (values) are accessed based on their **names** (keys)

**Example:** Suppose we want to work with the ages of people: Bob is 25, Ted is 32, Carol is 27, Alice is 28.

**Using a list:**
```python
ages_list = [25, 32, 27, 28]
```
This works, but we need to remember whose age is where. Ted's age is `ages_list[2]`.

**Using a dictionary:**
```python
ages_dict = {'Bob': 25, 'Ted': 32, 'Carol': 27, 'Alice': 28}
```
Now we can find Ted's age with `ages_dict['Ted']` - much clearer!

### Dictionary keys and values

You can get all keys or all values from a dictionary using these methods:

* **`dict.keys()`** - returns all keys in the dictionary
* **`dict.values()`** - returns all values in the dictionary
* **`dict.items()`** - returns all key-value pairs as tuples
* **`key in dict`** - returns `True` if the key exists in the dictionary, `False` otherwise

**Note:** In the following examples, we use the `list()` function to convert the results into actual list objects for easier viewing.

````python
# Run this cell to see dictionary keys and values in action

car = {"brand": "Toyota", "model": "Camry", "year": 2020, "color": "blue"}
print("Car dictionary:", car)

# Accessing values by key
print("Brand:", car["brand"])
print("Year:", car["year"])

# Getting all keys, values, and items
print("All keys:", list(car.keys()))
print("All values:", list(car.values()))
print("All items:", list(car.items()))

# Checking if a key exists
print("Has 'model' key:", "model" in car)
print("Has 'price' key:", "price" in car)

# Adding a new key-value pair
car["price"] = 25000
print("After adding price:", car)
````

### Common Errors: IndexError and KeyError

When working with lists and dictionaries, you'll encounter two common types of errors:

* **IndexError** - occurs when trying to access a list index that doesn't exist
* **KeyError** - occurs when trying to access a dictionary key that doesn't exist

Understanding these errors helps you debug your code and write more robust programs.

### Adding key-value pairs to dictionaries

Once we've defined a dictionary, it's easy to **add a key-value pair**. We just use an assignment statement, assigning the value to a new key:

**Example:**
```python
dct = {"a": 1, "b": 2}
dct["c"] = 3  # adds new key-value pair
print(dct)  # outputs: {"a": 1, "b": 2, "c": 3}
```

We can also **modify existing values** in the same way:
```python
dct["a"] = 10  # changes existing value
print(dct)  # outputs: {"a": 10, "b": 2, "c": 3}
```

**Key point:** If the key already exists, we'll update its value. If it doesn't exist, we'll create a new key-value pair.

### Typing long lists and dictionaries

It's hard to read code that has very long lines, because we have to use a scroll bar. Fortunately, there are ways to break lines when creating lists and dictionaries.

**Rule:** We can break anywhere **except before a comma**.

Code can be easier to read when items are stacked as in the second example below, though this is not commonly done for small dictionaries:

````python
inventory1 = {"apples": 50, "bananas": 30, "oranges": 25}

inventory2 = {
    "apples": 50,
    "bananas": 30,
    "oranges": 25
}

print(inventory1 == inventory2)
````

### Lists vs. Dictionaries: When to Use Which?

| Feature | Lists | Dictionaries |
|---------|-------|------------|
| **Structure** | Ordered sequence of elements | Unordered collection of key-value pairs |
| **Access Method** | By index (position): `my_list[0]` | By key (name): `my_dict["key"]` |
| **Indexing** | Numeric (0, 1, 2, ...) | Custom keys (strings, numbers, etc.) |
| **Order** | Maintains insertion order | Order not guaranteed (Python 3.7+ maintains insertion order) |
| **Duplicates** | Allows duplicate values | Keys must be unique, values can duplicate |
| **Best Used For** | - Sequences of similar items<br>- When order matters<br>- Mathematical operations | - Key-value relationships<br>- Looking up by name/ID<br>- Configuration settings |
| **Example Use Cases** | - Shopping list<br>- Test scores<br>- Time series data | - Student records<br>- Phone book<br>- Inventory with product codes |

### Tuples: Immutable Sequences

**Tuples** are another type of sequence in Python, similar to lists but with one key difference: they are **immutable**, meaning you cannot change their contents after creation.

**Key features:**
* Created using **parentheses** `()` instead of square brackets
* Items separated by commas
* Can also create a tuple without parentheses by just separating values with commas
* **Cannot be changed** after creation (immutable)

We'll see tuples when we discuss Python functions, as they're particularly useful for returning multiple values.

### The `len` function

`len` is short for length.  It returns the length of a string, list, dictionary, or tuple.

````python
string = "abc"
print(f"The length of the string is {len(string)}.")

lst = [1, 2, 3, "a", "b", "c"]
print(f"The length of the list is {len(lst)}.")

dct = {"a": 1, "b": 2, "c": 3, "d": 4, "e": 5}
print(f"The length of the dictionary is {len(dct)}.")

tple = (1, 2, "a", "b")
print(f"The length of the tuple is {len(tple)}.")
````
