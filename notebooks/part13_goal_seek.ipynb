## Part 1: Goal Seek in Python (Excel Goal Seek Equivalent)

Excel's Goal Seek changes one input cell to achieve a desired value in a formula cell. Python's `fsolve` does exactly the same thing!

**Excel Goal Seek:**
- Set cell: The formula cell you want to reach a target value
- To value: Your target
- By changing cell: The input to adjust

**Python fsolve:**
- Define an equation that equals zero at your target
- fsolve finds the input that makes it zero

Let's see this with a familiar example:

```python
# Example 1: Loan Payment Calculation (Classic Goal Seek Problem)
# Question: What interest rate gives me a $1,500 monthly payment?

def loan_payment_goal_seek():
    """
    Find the interest rate that results in a specific monthly payment.
    This is exactly what you'd use Excel Goal Seek for!
    """
    print("LOAN PAYMENT GOAL SEEK (Excel Goal Seek Equivalent)")
    print("="*60)
    
    # Loan parameters
    principal = 250000  # Loan amount
    years = 30          # Loan term
    target_payment = 1500  # Desired monthly payment
    
    def monthly_payment(annual_rate):
        """Calculate monthly payment for a given interest rate"""
        if annual_rate <= 0:
            return principal / (years * 12)
        
        monthly_rate = annual_rate / 12
        n_payments = years * 12
        
        # Standard loan payment formula (PMT function in Excel)
        payment = principal * (monthly_rate * (1 + monthly_rate)**n_payments) / \
                 ((1 + monthly_rate)**n_payments - 1)
        return payment
    
    def goal_seek_equation(rate):
        """This equals zero when payment equals our target"""
        return monthly_payment(rate) - target_payment
    
    print(f"Loan Details:")
    print(f"  Principal: ${principal:,}")
    print(f"  Term: {years} years")
    print(f"  Target Monthly Payment: ${target_payment:,}")
    
    # In Excel: Goal Seek on payment cell, target=$1,500, changing rate cell
    # In Python: fsolve finds the rate
    initial_guess = 0.05  # Start with 5% guess
    
    optimal_rate = fsolve(goal_seek_equation, initial_guess)[0]
    actual_payment = monthly_payment(optimal_rate)
    
    print(f"\nGoal Seek Results:")
    print(f"  Required Interest Rate: {optimal_rate:.3%}")
    print(f"  Actual Monthly Payment: ${actual_payment:.2f}")
    print(f"  Total Interest Paid: ${(actual_payment * years * 12 - principal):,.0f}")
    
    # Show what this looks like in Excel terms
    print(f"\nExcel Equivalent:")
    print(f"  Set cell: B5 (Monthly Payment) = ${target_payment}")
    print(f"  By changing cell: B2 (Interest Rate) → {optimal_rate:.3%}")
    
    return optimal_rate

# Run the goal seek example
interest_rate = loan_payment_goal_seek()
```

```python
# Example 2: Break-Even Analysis
# Question: How many units must I sell to break even?

def breakeven_goal_seek():
    """
    Find the sales volume needed to break even.
    Another classic Excel Goal Seek problem!
    """
    print("\nBREAK-EVEN ANALYSIS (Goal Seek)")
    print("="*60)
    
    # Business parameters
    fixed_costs = 50000
    variable_cost = 25
    selling_price = 45
    
    def profit(quantity):
        """Calculate profit for given quantity"""
        revenue = selling_price * quantity
        total_cost = fixed_costs + variable_cost * quantity
        return revenue - total_cost
    
    # Goal: Find quantity where profit = 0
    initial_guess = 1000
    breakeven_qty = fsolve(profit, initial_guess)[0]
    
    print(f"Business Setup:")
    print(f"  Fixed Costs: ${fixed_costs:,}")
    print(f"  Variable Cost per Unit: ${variable_cost}")
    print(f"  Selling Price: ${selling_price}")
    print(f"  Contribution Margin: ${selling_price - variable_cost} per unit")
    
    print(f"\nGoal Seek Results:")
    print(f"  Break-even Quantity: {breakeven_qty:,.0f} units")
    print(f"  Break-even Revenue: ${breakeven_qty * selling_price:,.0f}")
    
    # Verify
    print(f"\nVerification:")
    print(f"  Profit at break-even: ${profit(breakeven_qty):.2f} ✓")
    
    # Compare with formula
    formula_breakeven = fixed_costs / (selling_price - variable_cost)
    print(f"  Formula result: {formula_breakeven:,.0f} units")
    
    return breakeven_qty

# Run break-even analysis
breakeven = breakeven_goal_seek()
```

## Part 2: Multi-Variable Optimization (Excel Solver Equivalent)

Excel Solver can change multiple cells to optimize an objective while respecting constraints. Python's `minimize` function does the same but can handle much larger and more complex problems.

**Excel Solver Components:**
- Objective cell: Minimize or maximize
- Variable cells: What to change
- Constraints: Limits and requirements

**Python minimize Components:**
- Objective function: Function to minimize
- Variables: Array of decision variables
- Constraints: Dictionary of constraint functions

Let's see a familiar business problem:

```python
# Example 3: Product Mix Optimization (Classic Solver Problem)
# Maximize profit subject to resource constraints

def product_mix_optimization():
    """
    Optimize production quantities to maximize profit.
    This is the textbook Excel Solver problem!
    """
    print("PRODUCT MIX OPTIMIZATION (Excel Solver Equivalent)")
    print("="*60)
    
    # Product data
    products = ['Product A', 'Product B', 'Product C']
    profits = np.array([20, 30, 25])  # Profit per unit
    
    # Resource requirements per unit
    labor_required = np.array([2, 3, 2.5])  # Hours per unit
    material_required = np.array([3, 2, 4])  # Units per unit
    
    # Available resources
    labor_available = 500      # Total labor hours
    material_available = 600    # Total material units
    
    # Maximum demand
    max_demand = np.array([100, 120, 80])
    
    print("Problem Setup:")
    print("\nProducts:")
    for i, product in enumerate(products):
        print(f"  {product}: ${profits[i]} profit, {labor_required[i]}h labor, "
              f"{material_required[i]} materials, max {max_demand[i]} units")
    
    print(f"\nResources Available:")
    print(f"  Labor: {labor_available} hours")
    print(f"  Materials: {material_available} units")
    
    # Define optimization problem
    def objective(quantities):
        """Negative profit (we minimize, so negate for maximization)"""
        return -np.dot(quantities, profits)
    
    def labor_constraint(quantities):
        """Labor used must be <= available"""
        return labor_available - np.dot(quantities, labor_required)
    
    def material_constraint(quantities):
        """Materials used must be <= available"""
        return material_available - np.dot(quantities, material_required)
    
    # Set up constraints and bounds
    constraints = [
        {'type': 'ineq', 'fun': labor_constraint},    # >= 0
        {'type': 'ineq', 'fun': material_constraint}  # >= 0
    ]
    
    # Bounds: 0 <= quantity <= max_demand
    bounds = [(0, max_demand[i]) for i in range(len(products))]
    
    # Initial guess
    initial_quantities = np.array([50, 50, 40])
    
    # Solve (This is what Excel Solver does!)
    result = minimize(objective, initial_quantities, 
                     method='SLSQP',
                     bounds=bounds, 
                     constraints=constraints)
    
    if result.success:
        optimal_quantities = result.x
        optimal_profit = -result.fun
        
        print("\n" + "="*60)
        print("OPTIMAL SOLUTION:")
        print("="*60)
        
        for i, product in enumerate(products):
            print(f"{product}: Produce {optimal_quantities[i]:.0f} units")
        
        print(f"\nTotal Profit: ${optimal_profit:,.2f}")
        
        # Resource utilization
        labor_used = np.dot(optimal_quantities, labor_required)
        material_used = np.dot(optimal_quantities, material_required)
        
        print(f"\nResource Utilization:")
        print(f"  Labor: {labor_used:.0f}/{labor_available} hours "
              f"({labor_used/labor_available:.0%} utilized)")
        print(f"  Materials: {material_used:.0f}/{material_available} units "
              f"({material_used/material_available:.0%} utilized)")
        
        # Excel Solver equivalent setup
        print(f"\nExcel Solver Equivalent:")
        print(f"  Objective: Maximize cell D10 (Total Profit)")
        print(f"  By changing: B2:B4 (Production Quantities)")
        print(f"  Constraints:")
        print(f"    - E7 <= {labor_available} (Labor constraint)")
        print(f"    - E8 <= {material_available} (Material constraint)")
        print(f"    - B2:B4 >= 0 (Non-negative)")
        print(f"    - B2:B4 <= Max Demand")
        
        return optimal_quantities
    else:
        print(f"Optimization failed: {result.message}")
        return None

# Run product mix optimization
optimal_mix = product_mix_optimization()
```

```python
# Example 4: Portfolio Optimization
# Minimize risk for a target return (Modern Portfolio Theory)

def portfolio_optimization():
    """
    Find optimal portfolio weights to minimize risk.
    This is a more advanced Solver problem.
    """
    print("\nPORTFOLIO OPTIMIZATION (Advanced Solver)")
    print("="*60)
    
    # Asset data
    assets = ['Stocks', 'Bonds', 'Real Estate', 'Commodities']
    expected_returns = np.array([0.10, 0.04, 0.08, 0.06])  # Annual returns
    
    # Risk matrix (covariance)
    covariance = np.array([
        [0.04, 0.01, 0.02, 0.01],
        [0.01, 0.01, 0.005, 0.002],
        [0.02, 0.005, 0.03, 0.01],
        [0.01, 0.002, 0.01, 0.025]
    ])
    
    target_return = 0.07  # 7% target return
    
    print("Asset Information:")
    for i, asset in enumerate(assets):
        risk = np.sqrt(covariance[i, i])
        print(f"  {asset}: {expected_returns[i]:.1%} return, {risk:.1%} volatility")
    
    print(f"\nTarget Portfolio Return: {target_return:.1%}")
    
    def portfolio_risk(weights):
        """Calculate portfolio variance (risk squared)"""
        return np.dot(weights.T, np.dot(covariance, weights))
    
    def return_constraint(weights):
        """Portfolio return must equal target"""
        return np.dot(weights, expected_returns) - target_return
    
    def weight_constraint(weights):
        """Weights must sum to 1 (100%)"""
        return np.sum(weights) - 1.0
    
    # Constraints
    constraints = [
        {'type': 'eq', 'fun': return_constraint},
        {'type': 'eq', 'fun': weight_constraint}
    ]
    
    # Bounds: No short selling (0 <= weight <= 1)
    bounds = [(0, 1) for _ in range(len(assets))]
    
    # Initial guess: equal weights
    initial_weights = np.array([0.25, 0.25, 0.25, 0.25])
    
    # Optimize
    result = minimize(portfolio_risk, initial_weights,
                     method='SLSQP',
                     bounds=bounds,
                     constraints=constraints)
    
    if result.success:
        optimal_weights = result.x
        portfolio_volatility = np.sqrt(result.fun)
        sharpe_ratio = target_return / portfolio_volatility
        
        print("\n" + "="*60)
        print("OPTIMAL PORTFOLIO:")
        print("="*60)
        
        for i, asset in enumerate(assets):
            print(f"{asset:12s}: {optimal_weights[i]:6.1%}")
        
        print(f"\nPortfolio Metrics:")
        print(f"  Expected Return: {target_return:.1%}")
        print(f"  Portfolio Risk (Volatility): {portfolio_volatility:.1%}")
        print(f"  Sharpe Ratio: {sharpe_ratio:.2f}")
        
        # Compare with equal weights
        equal_weights = np.array([0.25, 0.25, 0.25, 0.25])
        equal_return = np.dot(equal_weights, expected_returns)
        equal_risk = np.sqrt(np.dot(equal_weights.T, np.dot(covariance, equal_weights)))
        
        print(f"\nComparison with Equal Weights (25% each):")
        print(f"  Equal Weight Return: {equal_return:.1%}")
        print(f"  Equal Weight Risk: {equal_risk:.1%}")
        print(f"  Risk Reduction: {(equal_risk - portfolio_volatility)/equal_risk:.1%}")
        
        return optimal_weights
    else:
        print(f"Optimization failed: {result.message}")
        return None

# Run portfolio optimization
optimal_portfolio = portfolio_optimization()
```

## Part 3: Advanced Techniques - Beyond Excel

Now let's explore optimization capabilities that Excel Solver can't handle easily:
- Nonlinear systems with multiple equations
- Large-scale problems (hundreds of variables)
- Complex constraint relationships

```python
# Example 5: Capital Budgeting Analysis (NPV Break-Even)
# Find the annual sales needed to achieve NPV = 0

def capital_budgeting_npv():
    """
    Find the annual sales volume needed to make a project have NPV = 0.
    This is a classic capital budgeting analysis that would be complex in Excel!
    """
    print("CAPITAL BUDGETING ANALYSIS (Beyond Excel)")
    print("="*60)
    
    # Project parameters
    initial_investment = 500000  # Initial cash outlay
    project_life = 5            # Years
    discount_rate = 0.12        # Cost of capital (12%)
    
    # Operating parameters per unit
    selling_price = 50          # Revenue per unit
    variable_cost = 30          # Variable cost per unit
    fixed_costs = 80000         # Annual fixed costs
    
    # Depreciation (straight line)
    annual_depreciation = initial_investment / project_life
    tax_rate = 0.25            # 25% tax rate
    
    print("Project Parameters:")
    print(f"  Initial Investment: ${initial_investment:,}")
    print(f"  Project Life: {project_life} years")
    print(f"  Discount Rate: {discount_rate:.1%}")
    print(f"  Selling Price per Unit: ${selling_price}")
    print(f"  Variable Cost per Unit: ${variable_cost}")
    print(f"  Annual Fixed Costs: ${fixed_costs:,}")
    print(f"  Tax Rate: {tax_rate:.1%}")
    
    def calculate_npv(annual_units):
        """Calculate NPV for a given annual sales volume"""
        
        # Annual cash flows
        annual_revenue = annual_units * selling_price
        annual_var_costs = annual_units * variable_cost
        
        # EBITDA (Earnings before interest, taxes, depreciation, amortization)
        ebitda = annual_revenue - annual_var_costs - fixed_costs
        
        # EBIT (Earnings before interest and taxes)
        ebit = ebitda - annual_depreciation
        
        # After-tax operating income
        after_tax_income = ebit * (1 - tax_rate)
        
        # Annual cash flow (add back depreciation - it's a non-cash expense)
        annual_cash_flow = after_tax_income + annual_depreciation
        
        # Calculate NPV
        npv = -initial_investment  # Initial outlay
        
        for year in range(1, project_life + 1):
            npv += annual_cash_flow / (1 + discount_rate)**year
        
        return npv
    
    def npv_equation(annual_units):
        """Equation that equals zero when NPV = 0"""
        return calculate_npv(annual_units)
    
    print(f"\nFinding break-even sales volume (NPV = 0):")
    
    # Find break-even sales volume
    initial_guess = 10000  # Start with 10,000 units
    
    try:
        breakeven_units = fsolve(npv_equation, initial_guess)[0]
        
        # Calculate key metrics at break-even
        annual_revenue = breakeven_units * selling_price
        annual_var_costs = breakeven_units * variable_cost
        ebitda = annual_revenue - annual_var_costs - fixed_costs
        ebit = ebitda - annual_depreciation
        after_tax_income = ebit * (1 - tax_rate)
        annual_cash_flow = after_tax_income + annual_depreciation
        
        print(f"\n" + "="*60)
        print(f"BREAK-EVEN ANALYSIS RESULTS:")
        print(f"="*60)
        
        print(f"\nRequired Annual Sales Volume: {breakeven_units:,.0f} units")
        print(f"Required Annual Revenue: ${annual_revenue:,.0f}")
        print(f"\nAnnual Operating Metrics:")
        print(f"  Revenue: ${annual_revenue:,.0f}")
        print(f"  Variable Costs: ${annual_var_costs:,.0f}")
        print(f"  Fixed Costs: ${fixed_costs:,.0f}")
        print(f"  EBITDA: ${ebitda:,.0f}")
        print(f"  Depreciation: ${annual_depreciation:,.0f}")
        print(f"  EBIT: ${ebit:,.0f}")
        print(f"  After-tax Income: ${after_tax_income:,.0f}")
        print(f"  Annual Cash Flow: ${annual_cash_flow:,.0f}")
        
        # Verify NPV = 0
        verification_npv = calculate_npv(breakeven_units)
        print(f"\nVerification:")
        print(f"  NPV at break-even volume: ${verification_npv:,.2f} ✓")
        
        # Sensitivity analysis
        print(f"\nSensitivity Analysis:")
        test_volumes = [breakeven_units * 0.9, breakeven_units * 1.1]
        
        for volume in test_volumes:
            npv = calculate_npv(volume)
            variance = (volume - breakeven_units) / breakeven_units
            print(f"  At {variance:+.0%} volume ({volume:,.0f} units): NPV = ${npv:+,.0f}")
        
        # Compare different scenarios
        print(f"\nScenario Analysis:")
        scenarios = [
            ("Optimistic", 1.2),
            ("Base Case", 1.0), 
            ("Pessimistic", 0.8)
        ]
        
        for scenario_name, multiplier in scenarios:
            volume = breakeven_units * multiplier
            npv = calculate_npv(volume)
            revenue = volume * selling_price
            print(f"  {scenario_name}: {volume:,.0f} units, ${revenue:,.0f} revenue, NPV = ${npv:+,.0f}")
        
        # Excel comparison
        print(f"\nWhy this is challenging in Excel:")
        print(f"  - Multi-period cash flow projections")
        print(f"  - Complex tax calculations")
        print(f"  - NPV formulas with variable inputs")
        print(f"  - Would require multiple interconnected worksheets")
        print(f"  - Goal Seek limited to single input/output relationship")
        
        return breakeven_units
        
    except Exception as e:
        print(f"Error in calculation: {e}")
        print("Try adjusting parameters or initial guess")
        return None

# Run capital budgeting analysis
breakeven_sales = capital_budgeting_npv()
```
