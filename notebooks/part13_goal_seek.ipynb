````python
# Example 3: System of Nonlinear Market Equations
# Find equilibrium in two interconnected markets with nonlinear relationships

def nonlinear_market_equilibrium():
    """
    Solve a system of nonlinear equations for market equilibrium
    Market A and Market B affect each other through cross-price elasticity
    """
    
    print("\nNONLINEAR MARKET EQUILIBRIUM SYSTEM")
    print("="*50)
    
    def market_system(variables):
        """
        System of two nonlinear market equations
        variables = [price_A, quantity_A, price_B, quantity_B]
        """
        P_A, Q_A, P_B, Q_B = variables
        
        # Market A equations (supply and demand must be equal)
        # Demand A: Q_A = 100 - 2*P_A + 0.5*P_B (substitutes)
        demand_A = 100 - 2*P_A + 0.5*P_B
        
        # Supply A: Q_A = 10 + 0.5*P_A^1.2 (nonlinear increasing costs)
        supply_A = 10 + 0.5 * (P_A ** 1.2)
        
        # Market B equations  
        # Demand B: Q_B = 80 - 1.5*P_B + 0.3*P_A (substitutes)
        demand_B = 80 - 1.5*P_B + 0.3*P_A
        
        # Supply B: Q_B = 5 + 0.8*P_B^1.1 (nonlinear increasing costs)
        supply_B = 5 + 0.8 * (P_B ** 1.1)
        
        # System of equations (all should equal zero at equilibrium)
        eq1 = Q_A - demand_A      # Quantity A = Demand A
        eq2 = Q_A - supply_A      # Quantity A = Supply A  
        eq3 = Q_B - demand_B      # Quantity B = Demand B
        eq4 = Q_B - supply_B      # Quantity B = Supply B
        
        return [eq1, eq2, eq3, eq4]
    
    # Initial guess: [price_A, quantity_A, price_B, quantity_B]
    initial_guess = [10, 30, 12, 25]
    
    # Solve the system
    solution = fsolve(market_system, initial_guess)
    P_A_eq, Q_A_eq, P_B_eq, Q_B_eq = solution
    
    # Verify solution
    residuals = market_system(solution)
    max_residual = max(abs(r) for r in residuals)
    
    print("Market System Description:")
    print("Market A: Consumer electronics (nonlinear supply costs)")
    print("Market B: Mobile accessories (substitute goods)")
    print("Cross-price elasticity: Products are substitutes")
    
    print(f"\nEquilibrium Solution:")
    print(f"Market A - Price: ${P_A_eq:.2f}, Quantity: {Q_A_eq:.1f} units")
    print(f"Market B - Price: ${P_B_eq:.2f}, Quantity: {Q_B_eq:.1f} units")
    print(f"Maximum residual: {max_residual:.6f} (should be ≈ 0)")
    
    # Calculate market metrics
    revenue_A = P_A_eq * Q_A_eq
    revenue_B = P_B_eq * Q_B_eq
    total_revenue = revenue_A + revenue_B
    
    print(f"\nMarket Analysis:")
    print(f"Market A Revenue: ${revenue_A:,.0f}")
    print(f"Market B Revenue: ${revenue_B:,.0f}")
    print(f"Total Market Revenue: ${total_revenue:,.0f}")
    
    # Cross-price elasticity analysis
    print(f"\nCross-Price Elasticity Effects:")
    
    # Calculate demand at equilibrium prices
    demand_A_eq = 100 - 2*P_A_eq + 0.5*P_B_eq
    demand_B_eq = 80 - 1.5*P_B_eq + 0.3*P_A_eq
    
    print(f"If Market B price increases by 10%:")
    P_B_higher = P_B_eq * 1.1
    demand_A_higher = 100 - 2*P_A_eq + 0.5*P_B_higher
    change_A = (demand_A_higher - demand_A_eq) / demand_A_eq
    print(f"  Market A demand changes by {change_A:+.1%}")
    
    print(f"If Market A price increases by 10%:")
    P_A_higher = P_A_eq * 1.1  
    demand_B_higher = 80 - 1.5*P_B_eq + 0.3*P_A_higher
    change_B = (demand_B_higher - demand_B_eq) / demand_B_eq
    print(f"  Market B demand changes by {change_B:+.1%}")
    
    return solution

# Run nonlinear market equilibrium
equilibrium_solution = nonlinear_market_equilibrium()
````

````python
# Example 2: Internal Rate of Return (IRR) Calculation
# Find the discount rate that makes NPV equal to zero

def irr_calculation_example():
    """
    Calculate IRR for a project with irregular cash flows using fsolve
    """
    
    print("\nINTERNAL RATE OF RETURN CALCULATION")
    print("="*50)
    
    # Project cash flows: [Year 0, Year 1, Year 2, Year 3, Year 4, Year 5]
    cash_flows = np.array([-100000, 25000, 30000, 35000, 40000, 20000])
    years = np.arange(len(cash_flows))
    
    print("Project Cash Flows:")
    for i, cf in enumerate(cash_flows):
        if i == 0:
            print(f"Year {i} (Initial Investment): ${cf:,}")
        else:
            print(f"Year {i}: ${cf:,}")
    
    def npv_equation(rate):
        """
        NPV equation that should equal zero at the IRR
        NPV = sum of [cash_flow / (1 + rate)^year] for all years
        """
        if rate <= -1:  # Avoid division by zero or negative denominators
            return float('inf')
        
        npv = 0
        for year, cash_flow in enumerate(cash_flows):
            npv += cash_flow / (1 + rate) ** year
        
        return npv
    
    # Find IRR using fsolve (where NPV = 0)
    initial_guess = 0.10  # Start with 10% guess
    irr = fsolve(npv_equation, initial_guess)[0]
    
    # Verify the solution
    npv_at_irr = npv_equation(irr)
    
    print(f"\nIRR Calculation Results:")
    print(f"Internal Rate of Return (IRR): {irr:.2%}")
    print(f"NPV at IRR: ${npv_at_irr:.2f} (should be ≈ 0)")
    
    # Calculate NPV at different discount rates for comparison
    print(f"\nNPV Sensitivity Analysis:")
    test_rates = [0.05, 0.08, irr, 0.12, 0.15]
    
    for rate in test_rates:
        npv = npv_equation(rate)
        rate_label = f"{rate:.2%}" if rate != irr else f"{rate:.2%} (IRR)"
        print(f"At {rate_label:12s}: NPV = ${npv:+8,.0f}")
    
    # Decision criteria
    print(f"\nInvestment Decision Criteria:")
    cost_of_capital = 0.10  # Assume 10% cost of capital
    
    if irr > cost_of_capital:
        decision = "ACCEPT"
        npv_at_cost = npv_equation(cost_of_capital)
        print(f"Cost of Capital: {cost_of_capital:.1%}")
        print(f"IRR vs Cost of Capital: {irr:.2%} > {cost_of_capital:.1%}")
        print(f"NPV at {cost_of_capital:.1%}: ${npv_at_cost:,.0f}")
        print(f"Recommendation: {decision} - Project creates value")
    else:
        decision = "REJECT"
        print(f"Cost of Capital: {cost_of_capital:.1%}")
        print(f"IRR vs Cost of Capital: {irr:.2%} < {cost_of_capital:.1%}")
        print(f"Recommendation: {decision} - Project destroys value")
    
    return irr

# Run IRR calculation
project_irr = irr_calculation_example()
````

````python
# Example 1: Break-Even Analysis with Nonlinear Cost Structure
# Find break-even point where total revenue equals total cost

def breakeven_analysis_example():
    """
    Find break-even quantity for a business with nonlinear cost structure
    """
    
    print("BREAK-EVEN ANALYSIS WITH NONLINEAR COSTS")
    print("="*50)
    
    # Business parameters
    fixed_costs = 50000        # Fixed costs per period
    variable_cost_per_unit = 25  # Variable cost per unit  
    price_per_unit = 45        # Selling price per unit
    
    # Nonlinear cost component (economies of scale)
    # Total cost has efficiency gains at higher volumes
    def total_cost(quantity):
        """Total cost with economies of scale"""
        if quantity <= 0:
            return fixed_costs
        
        # Base costs plus efficiency factor (costs decrease per unit at higher volumes)
        efficiency_factor = 1 / (1 + 0.0001 * quantity)  # Economies of scale
        variable_costs = variable_cost_per_unit * quantity * efficiency_factor
        
        return fixed_costs + variable_costs
    
    def total_revenue(quantity):
        """Total revenue (linear in this example)"""
        return price_per_unit * quantity
    
    def profit_equation(quantity):
        """Profit function - should equal zero at break-even"""
        return total_revenue(quantity) - total_cost(quantity)
    
    # Find break-even point using fsolve
    initial_guess = 3000  # Starting guess for quantity
    breakeven_quantity = fsolve(profit_equation, initial_guess)[0]
    
    # Calculate financial metrics at break-even
    breakeven_revenue = total_revenue(breakeven_quantity)
    breakeven_cost = total_cost(breakeven_quantity)
    
    print(f"Business Parameters:")
    print(f"Fixed Costs: ${fixed_costs:,}")
    print(f"Variable Cost per Unit: ${variable_cost_per_unit}")
    print(f"Selling Price per Unit: ${price_per_unit}")
    print(f"Economies of Scale: Yes (cost efficiency improves with volume)")
    
    print(f"\nBreak-Even Analysis Results:")
    print(f"Break-Even Quantity: {breakeven_quantity:,.0f} units")
    print(f"Break-Even Revenue: ${breakeven_revenue:,.0f}")
    print(f"Break-Even Total Cost: ${breakeven_cost:,.0f}")
    print(f"Profit at Break-Even: ${profit_equation(breakeven_quantity):,.2f} (should be ≈ 0)")
    
    # Calculate unit economics at break-even
    avg_cost_per_unit = breakeven_cost / breakeven_quantity
    margin_per_unit = price_per_unit - avg_cost_per_unit
    
    print(f"\nUnit Economics at Break-Even:")
    print(f"Average Cost per Unit: ${avg_cost_per_unit:.2f}")
    print(f"Margin per Unit: ${margin_per_unit:.2f}")
    print(f"Margin Percentage: {margin_per_unit/price_per_unit:.1%}")
    
    # Show sensitivity analysis
    print(f"\nSensitivity Analysis:")
    test_quantities = [breakeven_quantity * 0.8, breakeven_quantity * 1.2]
    
    for qty in test_quantities:
        profit = profit_equation(qty)
        margin = (qty - breakeven_quantity) / breakeven_quantity
        print(f"At {margin:+.0%} quantity ({qty:,.0f} units): Profit = ${profit:+,.0f}")
    
    return breakeven_quantity

# Run break-even analysis
breakeven_qty = breakeven_analysis_example()
````

# Nonlinear Equation Solving with scipy.optimize.fsolve

The `scipy.optimize.fsolve` function finds solutions to nonlinear equations and systems of equations. Unlike linear systems, these problems often involve exponential, logarithmic, trigonometric, or polynomial relationships that frequently appear in business and economics.

**Key Features:**
- **Root Finding:** Solves f(x) = 0 for single or multiple variables
- **Nonlinear Systems:** Handles complex interactions between variables
- **Numerical Methods:** Uses advanced algorithms (hybrid Powell method by default)
- **Flexible Input:** Accepts both single equations and systems of equations

**When to Use fsolve:**
- **Break-even Analysis:** Finding intersection points of cost and revenue functions
- **Financial Modeling:** Solving for IRR, yield-to-maturity, option pricing
- **Economic Equilibrium:** Market clearing with nonlinear supply/demand curves
- **Growth Models:** Finding steady-state solutions in dynamic systems
- **Optimization Conditions:** Solving first-order conditions (where derivatives = 0)

**Syntax:**
```python
solution = fsolve(equations, initial_guess)
# where equations is a function that returns residuals (should be zero at solution)
```

**Business Advantage:** Many real-world business relationships are nonlinear, making fsolve essential for realistic modeling.

````python
# Error Handling and Troubleshooting in Optimization

def error_handling_examples():
    """
    Demonstrate common optimization errors and how to handle them
    """
    print("ERROR HANDLING AND TROUBLESHOOTING")
    print("="*50)
    
    # Example 1: Infeasible problem
    print("\nExample 1: Infeasible Optimization Problem")
    print("-" * 40)
    
    def infeasible_problem():
        """Portfolio optimization with impossible constraints"""
        returns = np.array([0.05, 0.03, 0.08])
        target_return = 0.15  # Impossible target (higher than best asset)
        
        def portfolio_risk(weights):
            # Simple risk model
            return np.sum(weights**2)
        
        def return_constraint(weights):
            return np.dot(weights, returns) - target_return
        
        def weight_constraint(weights):
            return np.sum(weights) - 1.0
        
        bounds = [(0, 1) for _ in range(len(returns))]
        constraints = [
            {'type': 'eq', 'fun': return_constraint},
            {'type': 'eq', 'fun': weight_constraint}
        ]
        
        initial_weights = np.array([0.33, 0.33, 0.34])
        
        try:
            result = minimize(portfolio_risk, initial_weights, method='SLSQP',
                            bounds=bounds, constraints=constraints)
            
            if result.success:
                print("✓ Solution found (unexpected!)")
                print(f"  Weights: {result.x}")
                print(f"  Achieved return: {np.dot(result.x, returns):.3f}")
            else:
                print(f"✗ Optimization failed as expected: {result.message}")
                print(f"  Status code: {result.status}")
                
                # Diagnose the problem
                print(f"  Diagnosis:")
                print(f"    Target return: {target_return:.1%}")
                print(f"    Maximum possible return: {np.max(returns):.1%}")
                print(f"    Problem: Target exceeds maximum possible return")
                
                # Suggest solution
                feasible_target = np.max(returns) * 0.95
                print(f"  Suggested fix: Reduce target to {feasible_target:.1%}")
                
        except Exception as e:
            print(f"✗ Error during optimization: {e}")
    
    infeasible_problem()
    
    # Example 2: Poor initial guess
    print("\nExample 2: Poor Initial Guess")
    print("-" * 40)
    
    def poor_initial_guess():
        """Demonstrate how initial guess affects convergence"""
        
        def difficult_function(x):
            """Function with multiple local minima"""
            return x**4 - 4*x**3 + 4*x**2 + 1
        
        def function_equation(x):
            """Find where function equals specific value"""
            target_value = 2
            return difficult_function(x) - target_value
        
        initial_guesses = [-2, 0, 1, 3, 5]
        solutions = []
        
        print("Finding where f(x) = 2 for f(x) = x⁴ - 4x³ + 4x² + 1")
        print("Testing different initial guesses:")
        
        for guess in initial_guesses:
            try:
                solution = fsolve(function_equation, guess, full_output=True)
                x_solution = solution[0][0]
                info = solution[1]
                success = info['fvec'][0]
                
                # Verify solution
                actual_value = difficult_function(x_solution)
                error = abs(actual_value - 2)
                
                if error < 1e-6:
                    status = "✓ Good"
                    solutions.append(x_solution)
                else:
                    status = "✗ Poor"
                
                print(f"  Guess: {guess:4.1f} → Solution: {x_solution:6.3f}, "
                      f"f(x): {actual_value:6.3f}, Error: {error:.2e} {status}")
                
            except Exception as e:
                print(f"  Guess: {guess:4.1f} → Failed: {e}")
        
        # Show unique solutions found
        unique_solutions = []
        for sol in solutions:
            is_new = True
            for unique_sol in unique_solutions:
                if abs(sol - unique_sol) < 1e-3:
                    is_new = False
                    break
            if is_new:
                unique_solutions.append(sol)
        
        print(f"  Found {len(unique_solutions)} unique solutions: {[f'{s:.3f}' for s in unique_solutions]}")
        print(f"  Lesson: Different initial guesses can lead to different solutions")
    
    poor_initial_guess()
    
    # Example 3: Numerical issues
    print("\nExample 3: Numerical Stability Issues")
    print("-" * 40)
    
    def numerical_issues():
        """Handle numerical precision problems"""
        
        # Problem with extreme scaling
        def badly_scaled_problem():
            # Minimize x² + 1e12 * y²  (very different scales)
            def objective(vars):
                x, y = vars
                return x**2 + 1e12 * y**2
            
            initial_guess = [1e6, 1e-6]  # Poor scaling
            
            try:
                result = minimize(objective, initial_guess, method='BFGS')
                
                if result.success:
                    print("✓ Poorly scaled problem solved")
                    print(f"  Solution: x={result.x[0]:.2e}, y={result.x[1]:.2e}")
                    print(f"  Function value: {result.fun:.2e}")
                else:
                    print(f"✗ Poorly scaled problem failed: {result.message}")
                    
                    # Try with better scaling
                    print("  Attempting with scaling...")
                    def scaled_objective(vars):
                        x, y = vars
                        return (x/1e3)**2 + (y*1e6)**2
                    
                    result2 = minimize(scaled_objective, [1e3, 1e-12], method='BFGS')
                    if result2.success:
                        actual_x = result2.x[0] * 1e3
                        actual_y = result2.x[1] / 1e6
                        print(f"  ✓ Scaled version succeeded")
                        print(f"    Solution: x={actual_x:.2e}, y={actual_y:.2e}")
                
            except Exception as e:
                print(f"✗ Exception in badly scaled problem: {e}")
        
        badly_scaled_problem()
    
    numerical_issues()
    
    # Example 4: Robust wrapper function
    print("\nExample 4: Robust Optimization Wrapper")
    print("-" * 40)
    
    def robust_optimize(objective, initial_guess, **kwargs):
        """
        Robust optimization wrapper with error handling and multiple attempts
        """
        methods = kwargs.get('methods', ['SLSQP', 'trust-constr', 'L-BFGS-B'])
        max_attempts = kwargs.get('max_attempts', 3)
        
        results = []
        
        for method in methods:
            for attempt in range(max_attempts):
                try:
                    # Add some noise to initial guess for subsequent attempts
                    if attempt > 0:
                        noise_scale = 0.1 * attempt
                        noisy_guess = np.array(initial_guess) * (1 + noise_scale * np.random.randn(len(initial_guess)))
                    else:
                        noisy_guess = initial_guess
                    
                    result = minimize(objective, noisy_guess, method=method, **kwargs)
                    
                    if result.success:
                        results.append({
                            'method': method,
                            'attempt': attempt + 1,
                            'result': result,
                            'objective_value': result.fun
                        })
                        break  # Success, move to next method
                        
                except Exception as e:
                    print(f"  Method {method}, Attempt {attempt + 1}: Failed with {e}")
        
        if results:
            # Return best result
            best_result = min(results, key=lambda x: x['objective_value'])
            print(f"✓ Best solution found using {best_result['method']} (attempt {best_result['attempt']})")
            print(f"  Objective value: {best_result['objective_value']:.6f}")
            return best_result['result']
        else:
            print("✗ All optimization attempts failed")
            return None
    
    # Test robust wrapper
    def test_objective(x):
        return (x[0] - 1)**2 + (x[1] - 2)**2
    
    print("Testing robust optimization wrapper:")
    result = robust_optimize(test_objective, [10, -5], methods=['SLSQP', 'BFGS'])
    
    if result:
        print(f"  Final solution: x={result.x[0]:.3f}, y={result.x[1]:.3f}")
        print(f"  Expected solution: x=1.000, y=2.000")

# Run error handling examples
error_handling_examples()
````

### Error Handling and Troubleshooting

Optimization problems don't always have solutions or may fail to converge. Here's how to handle common issues and implement robust error handling.

````python
# Python vs Excel Goal Seek Comparison

def excel_vs_python_comparison():
    """
    Compare Excel Goal Seek with Python's optimization capabilities
    using a loan payment calculation example
    """
    print("EXCEL vs PYTHON GOAL SEEK COMPARISON")
    print("="*60)
    
    # Loan scenario: Find interest rate for desired monthly payment
    principal = 250000  # Loan amount
    years = 30  # Loan term
    desired_payment = 1500  # Target monthly payment
    
    def monthly_payment(annual_rate):
        """Calculate monthly payment given annual interest rate"""
        if annual_rate <= 0:
            return principal / (years * 12)  # Simple division if no interest
        
        monthly_rate = annual_rate / 12
        num_payments = years * 12
        
        # Standard loan payment formula
        payment = principal * (monthly_rate * (1 + monthly_rate)**num_payments) / \
                 ((1 + monthly_rate)**num_payments - 1)
        
        return payment
    
    def payment_equation(rate):
        """Equation that equals zero when payment equals target"""
        return monthly_payment(rate) - desired_payment
    
    print("Loan Scenario:")
    print(f"Principal: ${principal:,}")
    print(f"Term: {years} years ({years*12} payments)")
    print(f"Desired Monthly Payment: ${desired_payment:,}")
    
    # Method 1: Python fsolve (equivalent to Excel Goal Seek)
    print(f"\nMethod 1: Python fsolve (Excel Goal Seek equivalent)")
    initial_guess = 0.05  # Start with 5%
    
    try:
        optimal_rate = fsolve(payment_equation, initial_guess)[0]
        calculated_payment = monthly_payment(optimal_rate)
        
        print(f"✓ Solution found:")
        print(f"  Required Interest Rate: {optimal_rate:.4%}")
        print(f"  Calculated Payment: ${calculated_payment:.2f}")
        print(f"  Target Payment: ${desired_payment:.2f}")
        print(f"  Error: ${abs(calculated_payment - desired_payment):.2f}")
        
    except Exception as e:
        print(f"✗ Failed: {e}")
    
    # Method 2: Manual iteration (what Excel Goal Seek does internally)
    print(f"\nMethod 2: Manual iteration (Excel's internal process)")
    rate = 0.05  # Starting guess
    tolerance = 0.01  # $0.01 tolerance
    max_iterations = 100
    
    print(f"Iteration process (showing first 10 steps):")
    for i in range(max_iterations):
        current_payment = monthly_payment(rate)
        error = current_payment - desired_payment
        
        if i < 10:  # Show first 10 iterations
            print(f"  {i+1:2d}: Rate={rate:.4%}, Payment=${current_payment:7.2f}, Error=${error:+7.2f}")
        
        if abs(error) < tolerance:
            print(f"  ... Converged after {i+1} iterations")
            break
        
        # Numerical derivative (finite difference)
        delta_rate = 0.0001
        payment_plus = monthly_payment(rate + delta_rate)
        derivative = (payment_plus - current_payment) / delta_rate
        
        # Newton-Raphson update
        if abs(derivative) > 1e-10:
            rate = rate - error / derivative
        else:
            rate = rate * 1.1  # Simple adjustment if derivative is too small
    
    print(f"Manual iteration result: {rate:.4%}")
    
    # Compare capabilities
    print(f"\nCAPABILITY COMPARISON")
    print(f"="*60)
    
    comparison_data = {
        'Feature': [
            'Single variable goal seek',
            'Multiple variables',
            'Complex constraints', 
            'System of equations',
            'Global optimization',
            'Sensitivity analysis',
            'Solution validation',
            'Speed (complex problems)',
            'Ease of use (simple problems)',
            'Documentation/auditability',
            'Integration with analysis',
            'Cost'
        ],
        'Excel Goal Seek': [
            'Excellent', 'No', 'Limited', 'No', 'No', 'Manual', 
            'Limited', 'Fast', 'Excellent', 'Screenshots', 'Good', 'License required'
        ],
        'Python SciPy': [
            'Excellent', 'Yes', 'Excellent', 'Yes', 'Yes', 'Automated',
            'Excellent', 'Very fast', 'Moderate', 'Code + results', 'Excellent', 'Free'
        ]
    }
    
    comparison_df = pd.DataFrame(comparison_data)
    
    print("Feature Comparison:")
    for i, row in comparison_df.iterrows():
        print(f"{row['Feature']:25s}: Excel: {row['Excel Goal Seek']:15s} | Python: {row['Python SciPy']}")
    
    print(f"\nWhen to use Excel Goal Seek:")
    print(f"• Simple, one-time analysis")
    print(f"• Single variable problems")
    print(f"• Team members only know Excel")
    print(f"• Quick 'what-if' scenarios")
    print(f"• Results need to be embedded in spreadsheet")
    
    print(f"\nWhen to use Python:")
    print(f"• Complex optimization problems")
    print(f"• Multiple variables and constraints")
    print(f"• Repeated analysis/automation")
    print(f"• Need solution validation")
    print(f"• Integration with other analyses")
    print(f"• Large-scale or global optimization")
    
    return optimal_rate

# Run comparison
rate_comparison = excel_vs_python_comparison()
````

````python
# Solution Validation and Convergence Checking

def validation_example():
    """
    Demonstrate how to validate optimization solutions and check convergence
    """
    print("SOLUTION VALIDATION AND CONVERGENCE ANALYSIS")
    print("="*50)
    
    # Example: Portfolio optimization with validation
    returns = np.array([0.12, 0.08, 0.15, 0.06])  # Expected returns
    target_return = 0.10
    
    # Risk matrix (covariance)
    risk_matrix = np.array([
        [0.05, 0.02, 0.03, 0.01],
        [0.02, 0.02, 0.01, 0.005],
        [0.03, 0.01, 0.06, 0.02],
        [0.01, 0.005, 0.02, 0.01]
    ])
    
    def portfolio_risk(weights):
        """Portfolio variance (to minimize)"""
        return np.dot(weights.T, np.dot(risk_matrix, weights))
    
    def return_constraint(weights):
        """Return constraint: portfolio return = target"""
        return np.dot(weights, returns) - target_return
    
    def weight_constraint(weights):
        """Weight constraint: sum = 1"""
        return np.sum(weights) - 1.0
    
    # Set up optimization with detailed output
    bounds = [(0, 1) for _ in range(len(returns))]
    constraints = [
        {'type': 'eq', 'fun': return_constraint},
        {'type': 'eq', 'fun': weight_constraint}
    ]
    
    initial_weights = np.array([0.25, 0.25, 0.25, 0.25])
    
    # Solve with different methods to compare
    methods = ['SLSQP', 'trust-constr']
    results = {}
    
    for method in methods:
        print(f"\nOptimizing with method: {method}")
        result = minimize(portfolio_risk, initial_weights, method=method,
                         bounds=bounds, constraints=constraints,
                         options={'disp': False, 'maxiter': 1000})
        
        results[method] = result
        
        # Check convergence
        if result.success:
            print(f"✓ Optimization successful")
            print(f"  Iterations: {result.nit if hasattr(result, 'nit') else 'N/A'}")
            print(f"  Function evaluations: {result.nfev}")
            print(f"  Final objective value: {result.fun:.6f}")
            
            # Validate constraints
            weights = result.x
            achieved_return = np.dot(weights, returns)
            weight_sum = np.sum(weights)
            
            print(f"  Constraint validation:")
            print(f"    Target return: {target_return:.3f}")
            print(f"    Achieved return: {achieved_return:.3f}")
            print(f"    Return error: {abs(achieved_return - target_return):.6f}")
            print(f"    Weight sum: {weight_sum:.6f}")
            print(f"    Weight sum error: {abs(weight_sum - 1.0):.6f}")
            
            # Check if weights are reasonable
            if np.any(weights < -1e-6) or np.any(weights > 1.001):
                print(f"  ⚠ Warning: Weights outside expected bounds")
            
            # Gradient check (for unconstrained problems)
            try:
                from scipy.optimize import check_grad
                def objective_only(w):
                    return portfolio_risk(w)
                
                def objective_grad(w):
                    return 2 * np.dot(risk_matrix, w)
                
                grad_error = check_grad(objective_only, objective_grad, weights)
                print(f"  Gradient accuracy: {grad_error:.2e} (should be < 1e-5)")
            except:
                print(f"  Gradient check: Not available")
                
        else:
            print(f"✗ Optimization failed: {result.message}")
            print(f"  Status: {result.status}")
    
    # Compare solutions between methods
    if len(results) > 1:
        print(f"\nMethod Comparison:")
        for method, result in results.items():
            if result.success:
                risk = result.fun
                weights = result.x
                print(f"{method:15s}: Risk = {risk:.6f}, Weights = {weights}")
    
    # Sensitivity analysis
    print(f"\nSensitivity Analysis:")
    if 'SLSQP' in results and results['SLSQP'].success:
        base_weights = results['SLSQP'].x
        base_risk = results['SLSQP'].fun
        
        # Test small perturbations
        for i, asset in enumerate(['Stocks', 'Bonds', 'REITs', 'Cash']):
            perturbed_weights = base_weights.copy()
            perturbed_weights[i] += 0.01  # 1% increase
            perturbed_weights = perturbed_weights / np.sum(perturbed_weights)  # Renormalize
            
            new_risk = portfolio_risk(perturbed_weights)
            risk_change = new_risk - base_risk
            
            print(f"  +1% {asset:6s}: Risk change = {risk_change:+.6f}")
    
    return results

# Run validation example
validation_results = validation_example()
````

````python
# Example 1: Multi-Market Equilibrium Analysis
# Find equilibrium prices and quantities for interconnected markets

def market_equilibrium_example():
    """
    Solve for equilibrium in three interconnected markets:
    - Market A: Consumer goods (affected by markets B and C)
    - Market B: Raw materials (affects market A)  
    - Market C: Labor (affects both markets A and B)
    """
    
    print("MULTI-MARKET EQUILIBRIUM ANALYSIS")
    print("="*50)
    
    # System of equations representing market relationships:
    # Market A: 2*P_A - 0.5*P_B - 0.3*P_C = 10  (consumer goods)
    # Market B: -0.4*P_A + 3*P_B - 0.2*P_C = 8   (raw materials)
    # Market C: -0.2*P_A - 0.3*P_B + 2.5*P_C = 6 (labor)
    # Where P_A, P_B, P_C are equilibrium prices
    
    # Coefficient matrix A
    A = np.array([
        [ 2.0, -0.5, -0.3],  # Market A equation
        [-0.4,  3.0, -0.2],  # Market B equation  
        [-0.2, -0.3,  2.5]   # Market C equation
    ])
    
    # Constants vector b
    b = np.array([10, 8, 6])
    
    # Solve the system
    equilibrium_prices = solve(A, b)
    
    print("Market System Equations:")
    print("Market A: 2.0*P_A - 0.5*P_B - 0.3*P_C = 10")
    print("Market B: -0.4*P_A + 3.0*P_B - 0.2*P_C = 8") 
    print("Market C: -0.2*P_A - 0.3*P_B + 2.5*P_C = 6")
    
    print(f"\nEquilibrium Solution:")
    markets = ['Consumer Goods (A)', 'Raw Materials (B)', 'Labor (C)']
    
    for i, market in enumerate(markets):
        price = equilibrium_prices[i]
        print(f"{market:18s}: ${price:7.2f}")
    
    # Verify solution by substituting back
    print(f"\nVerification (Ax should equal b):")
    verification = A @ equilibrium_prices
    
    for i in range(len(b)):
        print(f"Equation {i+1}: {verification[i]:.3f} = {b[i]:.3f} ✓")
    
    # Economic interpretation
    print(f"\nEconomic Interpretation:")
    print(f"• Consumer goods have the highest equilibrium price")
    print(f"• Raw materials and labor costs are balanced")
    print(f"• All markets clear simultaneously at these prices")
    
    return equilibrium_prices

# Run market equilibrium analysis
equilibrium_prices = market_equilibrium_example()
````

# Linear Systems with scipy.linalg.solve

The `scipy.linalg.solve` function efficiently solves systems of linear equations of the form Ax = b, where A is a matrix of coefficients, x is the vector of unknowns, and b is the vector of constants. This is fundamental for many business applications involving simultaneous relationships.

**Mathematical Form:**
```
A₁₁x₁ + A₁₂x₂ + ... + A₁ₙxₙ = b₁
A₂₁x₁ + A₂₂x₂ + ... + A₂ₙxₙ = b₂
...
Aₘ₁x₁ + Aₘ₂x₂ + ... + Aₘₙxₙ = bₘ
```

**Key Advantages:**
- **Exact Solutions:** Unlike optimization, linear systems have precise mathematical solutions
- **Computational Efficiency:** Highly optimized algorithms for large systems
- **Numerical Stability:** Robust handling of near-singular matrices
- **Versatility:** Handles square, overdetermined, and underdetermined systems

**Business Applications:**
- **Market Equilibrium:** Supply and demand balance across multiple markets
- **Input-Output Analysis:** Economic relationships between industry sectors
- **Financial Planning:** Budget allocation with multiple constraints
- **Network Flow:** Transportation and logistics optimization
- **Regression Analysis:** Solving normal equations for least squares

**Syntax:**
```python
solution = solve(A, b)  # Solves Ax = b for x
```

````python
# Intermediate Exercise: Sales Commission Structure
# Find the commission rate that achieves target sales performance

def sales_commission_exercise():
    """
    A company wants to design a commission structure that motivates salespeople
    to achieve exactly $500,000 in annual sales revenue per person.
    
    Commission structure: Base salary + commission rate × (sales - threshold)
    Sales motivation model: sales = base_performance × (1 + motivation_factor)
    Motivation factor depends on potential earnings above base salary.
    """
    print("SALES COMMISSION OPTIMIZATION EXERCISE")
    print("="*50)
    
    # Given parameters
    base_salary = 50000  # Annual base salary
    base_performance = 300000  # Sales without commission motivation
    threshold = 100000  # Minimum sales before commission kicks in
    target_sales = 500000  # Target sales per salesperson
    
    def sales_model(commission_rate):
        """
        Model how commission rate affects sales performance
        Higher commission rates increase motivation and sales
        """
        if commission_rate <= 0:
            return base_performance
        
        # Calculate potential commission earnings at target level
        potential_commission = max(0, commission_rate * (target_sales - threshold))
        
        # Motivation factor: higher potential earnings = higher motivation
        # But with diminishing returns (logarithmic relationship)
        motivation_factor = 0.5 * np.log(1 + potential_commission / base_salary)
        
        # Actual sales = base performance + motivation boost
        actual_sales = base_performance * (1 + motivation_factor)
        
        return actual_sales
    
    def sales_target_equation(commission_rate):
        """Equation that equals zero when actual sales = target sales"""
        return sales_model(commission_rate) - target_sales
    
    print("Commission Structure Design:")
    print(f"Base Salary: ${base_salary:,}")
    print(f"Sales Threshold: ${threshold:,}")
    print(f"Base Performance (no commission): ${base_performance:,}")
    print(f"Target Sales: ${target_sales:,}")
    print(f"Required Improvement: {(target_sales - base_performance) / base_performance:.1%}")
    
    # Solve for optimal commission rate
    initial_guess = 0.05  # Start with 5%
    
    try:
        optimal_commission = fsolve(sales_target_equation, initial_guess)[0]
        
        # Calculate results
        achieved_sales = sales_model(optimal_commission)
        commission_earned = max(0, optimal_commission * (achieved_sales - threshold))
        total_compensation = base_salary + commission_earned
        
        print(f"\nOptimal Commission Structure:")
        print(f"Commission Rate: {optimal_commission:.2%}")
        print(f"Achieved Sales: ${achieved_sales:,.0f}")
        print(f"Commission Earned: ${commission_earned:,.0f}")
        print(f"Total Compensation: ${total_compensation:,.0f}")
        
        # Verify solution
        error = abs(achieved_sales - target_sales)
        print(f"\nSolution Verification:")
        print(f"Target vs Achieved: ${target_sales:,} vs ${achieved_sales:,.0f}")
        print(f"Error: ${error:.0f} (should be close to 0)")
        
        # Sensitivity analysis
        print(f"\nSensitivity Analysis:")
        test_rates = [optimal_commission * 0.8, optimal_commission, optimal_commission * 1.2]
        
        for rate in test_rates:
            sales = sales_model(rate)
            commission = max(0, rate * (sales - threshold))
            total_comp = base_salary + commission
            variance = (sales - target_sales) / target_sales
            
            print(f"Rate {rate:.2%}: Sales ${sales:,.0f} ({variance:+.1%}), "
                  f"Total Comp ${total_comp:,.0f}")
        
        return optimal_commission
        
    except Exception as e:
        print(f"Error finding solution: {e}")
        print("Try adjusting the initial guess or model parameters")
        return None

# Run the exercise
optimal_commission = sales_commission_exercise()
````

````python
# Example: Simple Break-Even Analysis
# Goal: Find the sales volume needed to break even

def simple_breakeven_example():
    """
    Find break-even point for a simple linear cost structure
    This is equivalent to Excel's Goal Seek functionality
    """
    print("SIMPLE BREAK-EVEN ANALYSIS (Goal Seek)")
    print("="*50)
    
    # Business parameters
    fixed_costs = 25000
    variable_cost_per_unit = 15
    selling_price_per_unit = 35
    
    def profit_function(quantity):
        """Calculate profit for a given quantity"""
        revenue = selling_price_per_unit * quantity
        total_costs = fixed_costs + (variable_cost_per_unit * quantity)
        return revenue - total_costs
    
    def breakeven_equation(quantity):
        """Equation that equals zero at break-even point"""
        return profit_function(quantity)
    
    print(f"Business Parameters:")
    print(f"Fixed Costs: ${fixed_costs:,}")
    print(f"Variable Cost per Unit: ${variable_cost_per_unit}")
    print(f"Selling Price per Unit: ${selling_price_per_unit}")
    print(f"Contribution Margin: ${selling_price_per_unit - variable_cost_per_unit} per unit")
    
    # Goal seek: Find quantity where profit = 0
    initial_guess = 1000
    breakeven_qty = fsolve(breakeven_equation, initial_guess)[0]
    
    print(f"\nGoal Seek Results:")
    print(f"Break-even quantity: {breakeven_qty:,.0f} units")
    print(f"Break-even revenue: ${breakeven_qty * selling_price_per_unit:,.0f}")
    print(f"Verification - Profit at break-even: ${profit_function(breakeven_qty):.2f}")
    
    # Compare with analytical solution
    analytical_breakeven = fixed_costs / (selling_price_per_unit - variable_cost_per_unit)
    print(f"\nComparison with analytical solution: {analytical_breakeven:,.0f} units")
    print(f"Difference: {abs(breakeven_qty - analytical_breakeven):.6f} (should be ≈ 0)")
    
    return breakeven_qty

# Run simple break-even analysis
simple_breakeven = simple_breakeven_example()
````
