### Floats

Numbers with decimal points are considered a different type of object than integers, because they require more space in memory to store.  They are called **floating point** objects, or, more simply, **floats**. 

````python
# Run this cell. 

x = 10.0
print(type(x))
````

### Mathematical operations

We can perform standard mathematical objects on ints or floats or a combination of ints and floats.

* addition (+)
* subtraction (-)
* multiplication (*)
* exponentiation (**)

````python
# Run this cell. 

print("2*10.0 =", 2*10.0)
print("3**2 =", 3**2)  # ^ in Excel is ** in python
````

### Division operations

Python has three types of division operators:

* division (/) - returns a float even if dividing two integers
* floor division (//) - returns the largest integer less than or equal to the result
* modulo (%) - returns the remainder after division

````python
# Run this cell.

print("10 / 3 =", 10 / 3)        # regular division (float result)
print("10 // 3 =", 10 // 3)      # floor division (integer result)
print("10 % 3 =", 10 % 3)        # modulo (remainder)
print("15 / 5 =", 15 / 5)        # even division still returns float
````

### Type conversions

Python provides built-in functions to convert between different types:

* `int()` - converts to integer (truncates decimals, doesn't round)
* `float()` - converts to floating point number
* `str()` - converts to string
* `bool()` - converts to boolean (False for 0, empty strings, None; True otherwise)

````python
# Run this cell.

print("int(10.7) =", int(10.7))      # truncates to 10
print("float(5) =", float(5))        # converts to 5.0
print("str(42) =", str(42))          # converts to "42"
print("bool(1) =", bool(1))          # converts to True
print("bool(0) =", bool(0))          # converts to False
print("bool('') =", bool(''))        # empty string is False
print("bool('hello') =", bool('hello'))  # non-empty string is True
````

* An **attribute** is a value or property associated with an object.

To access an attribute or a method we type a period (`.`) after the object or the object's name and then type the attribute or method name. ### Defining new classes

So far we've worked with built-in Python objects like integers, strings, and booleans. But Python allows us to create our own custom types of objects by defining **classes**.

A **class** is like a blueprint or template for creating objects. It defines what attributes (data) and methods (functions) the objects will have.

The basic syntax for defining a class is:

```python
class ClassName:
    def __init__(self, parameters):
        # Initialize the object's attributes
        self.attribute = value
    
    def method_name(self, parameters):
        # Define what the method does
        return something
```

* `__init__` is a special method called the **constructor** - it runs when you create a new object
* `self` refers to the specific object instance being created or used

````python
# Run this cell to see examples of defining and using classes

# Define a simple Person class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def greet(self):
        return f"Hello, my name is {self.name} and I'm {self.age} years old."
    
    def have_birthday(self):
        self.age += 1
        return f"Happy birthday! {self.name} is now {self.age} years old."

# Create objects (instances) of the Person class
alice = Person("Alice", 25)
bob = Person("Bob", 30)

print("Alice's greeting:", alice.greet())
print("Bob's greeting:", bob.greet())

# Access attributes directly
print(f"Alice's name: {alice.name}")
print(f"Bob's age: {bob.age}")

# Call methods
print(alice.have_birthday())
print(f"Alice's age is now: {alice.age}")

# Show that objects are independent
print(f"Bob's age is still: {bob.age}")  # Bob didn't have a birthday
````

### More class examples

Here's another example showing a BankAccount class that demonstrates how classes can maintain state and provide useful functionality:

````python
# Run this cell to see a BankAccount class example

class BankAccount:
    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        self.balance = initial_balance
        self.transaction_history = []
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.transaction_history.append(f"Deposited ${amount}")
            return f"Deposited ${amount}. New balance: ${self.balance}"
        else:
            return "Deposit amount must be positive"
    
    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            self.transaction_history.append(f"Withdrew ${amount}")
            return f"Withdrew ${amount}. New balance: ${self.balance}"
        elif amount > self.balance:
            return "Insufficient funds"
        else:
            return "Withdrawal amount must be positive"
    
    def get_balance(self):
        return f"{self.owner}'s account balance: ${self.balance}"
    
    def get_history(self):
        return self.transaction_history

# Create and use bank accounts
account1 = BankAccount("Alice", 1000)
account2 = BankAccount("Bob")  # Uses default balance of 0

print(account1.get_balance())
print(account2.get_balance())

print("\nAlice's transactions:")
print(account1.withdraw(200))
print(account1.deposit(50))
print(account1.withdraw(2000))  # Should fail

print("\nTransaction history:")
for transaction in account1.get_history():
    print(f"- {transaction}")

print(f"\nFinal balance: {account1.get_balance()}")
````
