### If-Elif-Else Statements

**Conditional statements** allow your program to make decisions and execute different code based on conditions. Python uses `if`, `elif` (else if), and `else` statements to control program flow.

The basic syntax is:
```python
if condition:
    # code to execute if condition is True
elif another_condition:
    # code to execute if another_condition is True
else:
    # code to execute if no conditions are True
```

Key points:
- Conditions are evaluated as True or False (boolean values)
- Use comparison operators: `==`, `!=`, `<`, `>`, `<=`, `>=`
- Use logical operators: `and`, `or`, `not`
- Indentation is crucial - Python uses indentation to define code blocks
- You can use as many `elif`'s as you need

````python
# Run this cell to see if-elif-else in action

# Example 1: Simple if statement
age = 18
if age >= 18:
    print("You are an adult")

# Example 2: If-else statement
temperature = 75
if temperature > 80:
    print("It's hot outside")
else:
    print("It's not too hot")

# Example 3: If-elif-else chain
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(f"Score: {score}, Grade: {grade}")

# Example 4: Using comparison operators
x = 10
y = 5
if x > y:
    print(f"{x} is greater than {y}")
elif x < y:
    print(f"{x} is less than {y}")
else:
    print(f"{x} equals {y}")
````

### Logical Operators

Python provides logical operators to combine multiple conditions:

- `and` - True if both conditions are True
- `or` - True if at least one condition is True  
- `not` - True if the condition is False (negation)

These operators allow you to create more complex conditional logic.

````python
# Run this cell to see logical operators in action

# Example 1: Using 'and' operator
username = "admin"
password = "secret123"
if username == "admin" and password == "secret123":
    print("Login successful")
else:
    print("Invalid credentials")

# Example 2: Using 'or' operator
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a weekday")

# Example 3: Using 'not' operator
is_raining = False
if not is_raining:
    print("You don't need an umbrella")
else:
    print("Take an umbrella")

# Example 4: Complex conditions
age = 25
has_license = True
has_car = True

if age >= 18 and has_license and has_car:
    print("You can drive!")
elif age >= 18 and has_license:
    print("You can drive, but you need a car")
elif age >= 18:
    print("You can get a license")
else:
    print("You're too young to drive")

# Example 5: Membership testing
vowels = "aeiou"
letter = "a"
if letter in vowels:
    print(f"{letter} is a vowel")
else:
    print(f"{letter} is a consonant")
````

### Float precision and comparisons

When working with floating-point numbers and if statements, you need to be aware of precision issues. Floats are stored in binary format, which can lead to small precision errors.

A famous example is that `0.1 + 0.2` does not exactly equal `0.3` in floating-point arithmetic. This means that direct equality comparisons with floats can sometimes fail unexpectedly.

Instead of exact equality checks, you should either:
1. Use the `round()` function to compare rounded values
2. Check if the difference between two floats is very small (tolerance-based comparison)

````python
# Run this cell to see floating-point precision issues with if statements

print("Direct comparison problems:")
result = 0.1 + 0.2
print(f"0.1 + 0.2 = {result}")
print(f"0.1 + 0.2 == 0.3: {result == 0.3}")
print(f"Actual result: {repr(result)}")

print("\nSolution 1: Using round() function")
if round(0.1 + 0.2, 1) == round(0.3, 1):
    print("✓ Using round(1) - they are equal!")
else:
    print("✗ Still not equal")

if round(0.1 + 0.2, 10) == round(0.3, 10):
    print("✓ Using round(10) - they are equal!")
else:
    print("✗ Still not equal")

print("\nSolution 2: Tolerance-based comparison")
tolerance = 0.0001
if abs(0.1 + 0.2 - 0.3) < tolerance:
    print("✓ Within tolerance - close enough!")
else:
    print("✗ Not within tolerance")

print("\nPractical example - price calculations:")
price = 0.1 + 0.1 + 0.1  # Three items at $0.10 each
expected_total = 0.3

if round(price, 2) == round(expected_total, 2):
    print(f"✓ Price check passed: ${round(price, 2)}")
else:
    print(f"✗ Price mismatch: ${price} vs ${expected_total}")
````

### Compact If-Else (Ternary Operator)

Python provides a **compact way to write simple if-else statements** using the **ternary operator** (also called conditional expression). This allows you to write a simple if-else statement in a single line.

The syntax is:
```python
value_if_true if condition else value_if_false
```

This is equivalent to:
```python
if condition:
    result = value_if_true
else:
    result = value_if_false
```

The compact form is useful for simple assignments and when you need to choose between two values based on a condition. However, don't overuse it - if the logic becomes complex, regular if-else statements are more readable.

````python
# Run this cell s

# Regular way:
temperature = 85
if temperature > 80:
    clothing = "shorts"
else:
    clothing = "pants"
print(f"Regular if-else: {clothing}")

# Compact way:
clothing_compact = "shorts" if temperature > 80 else "pants"
print(f"Compact form: {clothing_compact}")
````

### For Loops

**For loops** allow us to iterate through a sequence of items (like lists, strings, ranges) and execute code for each item. They are essential for automating repetitive tasks.

Basic syntax:
```python
for item in sequence:
    # code to execute for each item
```

Common patterns:
* `for item in list:` - iterate through list items
* `for i in range(n):` - iterate n times with index
* `for i, item in enumerate(list):` - get both index and item
* `for key, value in dict.items():` - iterate through dictionary
* `for char in string:` - iterate through string characters

For loops can also use `break` to exit early and `continue` to skip to the next iteration.

### The range() function

The `range()` function is commonly used with for loops to generate sequences of numbers:

* `range(n)` - generates numbers from 0 to n-1
* `range(start, stop)` - generates numbers from start to stop-1  
* `range(start, stop, step)` - generates numbers from start to stop-1, incrementing by step

Examples:
* `range(5)` → 0, 1, 2, 3, 4
* `range(2, 8)` → 2, 3, 4, 5, 6, 7
* `range(0, 10, 2)` → 0, 2, 4, 6, 8

````python
# Run this cell to see for loops in action

# Example 1: Iterating through a list
fruits = ["apple", "banana", "cherry"]
print("Fruits:")
for fruit in fruits:
    print(f"- {fruit}")

# Example 2: Using range()
print("\nCounting to 5:")
for i in range(1, 6):
    print(f"Count: {i}")

# Example 3: Iterating with index using enumerate()
colors = ["red", "green", "blue"]
print("\nColors with indices:")
for i, color in enumerate(colors):
    print(f"Index {i}: {color}")

# Example 4: Iterating through a string
word = "Python"
print(f"\nLetters in '{word}':")
for letter in word:
    print(letter)

# Example 5: Iterating through a dictionary
student_grades = {"Alice": 85, "Bob": 92, "Charlie": 78}
print("\nStudent grades:")
for name, grade in student_grades.items():
    print(f"{name}: {grade}")
````

### While Loops

**While loops** repeatedly execute code as long as a condition remains true. Unlike for loops, which iterate over a known sequence, while loops continue until a condition becomes false.

Basic syntax:
```python
while condition:
    # code to execute while condition is True
    # usually includes code that will eventually make condition False
```

Key considerations:
* Make sure the condition will eventually become False to avoid infinite loops
* Often used when you don't know how many iterations you need
* Common to modify a counter or update a variable inside the loop
* Can use `break` to exit early and `continue` to skip to next iteration

While loops are useful for user input validation, game loops, and situations where you need to repeat until some condition is met.

````python
# Run this cell to see while loops in action

# Example 1: Basic while loop with counter
print("Countdown:")
count = 5
while count > 0:
    print(count)
    count -= 1  # Decrease count by 1
print("Blast off!")

# Example 2: Accumulating values
print("\nSum of numbers until we reach 50:")
total = 0
number = 1
while total < 50:
    total += number
    print(f"Added {number}, total now: {total}")
    number += 1
````

### Error Handling with Try/Except

**Error handling** allows your program to gracefully deal with unexpected situations instead of crashing. Python uses `try`/`except` blocks to catch and handle errors (called exceptions).

Basic syntax:
```python
try:
    # code that might cause an error
    risky_code()
except SpecificError:
    # handle specific error type
    print("Handle specific error")
except:
    # handle any other error
    print("Handle any error")
```

**Common exception types:**
- `ValueError` - invalid value for operation (e.g., `int("abc")`)
- `ZeroDivisionError` - division by zero
- `IndexError` - list index out of range  
- `KeyError` - dictionary key not found
- `FileNotFoundError` - file doesn't exist
- `TypeError` - wrong data type for operation

Error handling is essential for user input validation, file operations, and making robust programs.

````python
# Run this cell to see basic try/except in action

Handling division by zero
try:
    result = 10 / 0
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
````

### List and Dictionary Comprehensions

**Comprehensions** provide a concise way to create lists and dictionaries using a compact syntax. They combine loops and conditional logic into a single expression.

**List Comprehensions:**
```python
new_list = [expression for item in iterable if condition]
```

**Dictionary Comprehensions:**
```python
new_dict = {key_expr: value_expr for item in iterable if condition}
```

Comprehensions are often more readable and faster than equivalent for loops. They're particularly useful for transforming data, filtering collections, and creating new data structures from existing ones.

The `if condition` part is optional and can be used to filter which items to include.

````python
# Run this cell to see comprehensions in action

# List Comprehensions
print("=== LIST COMPREHENSIONS ===")

# Example 1: Square numbers
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]
print(f"Numbers: {numbers}")
print(f"Squares: {squares}")

# Example 2: Filter even numbers and double them
evens_doubled = [x * 2 for x in range(10) if x % 2 == 0]
print(f"Even numbers doubled: {evens_doubled}")


# Dictionary Comprehensions
print("\n=== DICTIONARY COMPREHENSIONS ===")

# Example 3: Create a dictionary of squares
square_dict = {x: x**2 for x in range(1, 6)}
print(f"Square dictionary: {square_dict}")

# Example 4: Transform existing dictionary
prices = {"apple": 1.20, "banana": 0.50, "orange": 0.80}
expensive_items = {item: price for item, price in prices.items() if price > 0.75}
print(f"Expensive items: {expensive_items}")

# Example 6: Create dictionary from two lists
fruits = ["apple", "banana", "cherry"]
colors = ["red", "yellow", "red"]
fruit_colors = {fruit: color for fruit, color in zip(fruits, colors)}
print(f"Fruit colors: {fruit_colors}")
````
