### Introduction to Flow Control

**Flow control** refers to the order in which your program executes code. By default, Python executes statements line by line from top to bottom. However, flow control statements allow you to:

- Make decisions with **conditional statements** (`if`, `elif`, `else`)
- Repeat code with **loops** (`for`, `while`)
- Create concise data transformations with **comprehensions**

These tools let you write programs that can respond to different inputs, process collections of data, and automate repetitive tasks.

### If-Elif-Else Statements

**Conditional statements** allow your program to make decisions and execute different code based on conditions. Python uses `if`, `elif` (else if), and `else` statements to control program flow.

The basic syntax is:
```python
if condition:
    # code to execute if condition is True
elif another_condition:
    # code to execute if another_condition is True
else:
    # code to execute if no conditions are True
```

Key points:
- Conditions are evaluated as True or False (boolean values)
- Use comparison operators: `==`, `!=`, `<`, `>`, `<=`, `>=`
- Use logical operators: `and`, `or`, `not`
- Indentation is crucial - Python uses indentation to define code blocks

````python
# Run this cell to see if-elif-else in action

# Example 1: Simple if statement
age = 18
if age >= 18:
    print("You are an adult")

# Example 2: If-else statement
temperature = 75
if temperature > 80:
    print("It's hot outside")
else:
    print("It's not too hot")

# Example 3: If-elif-else chain
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(f"Score: {score}, Grade: {grade}")

# Example 4: Using comparison operators
x = 10
y = 5
if x > y:
    print(f"{x} is greater than {y}")
elif x < y:
    print(f"{x} is less than {y}")
else:
    print(f"{x} equals {y}")
````

### Logical Operators

Python provides logical operators to combine multiple conditions:

- `and` - True if both conditions are True
- `or` - True if at least one condition is True

These operators allow you to create more complex conditional logic.

````python
# Run this cell to see logical operators in action

# Example 1: Using 'and' operator
username = "admin"
password = "secret123"
if username == "admin" and password == "secret123":
    print("Login successful")
else:
    print("Invalid credentials")

# Example 2: Using 'or' operator
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a weekday")

# Example 3: Using 'not' operator
is_raining = False
if not is_raining:
    print("You don't need an umbrella")
else:
    print("Take an umbrella")

# Example 4: Complex conditions
age = 25
has_license = True
has_car = True

if age >= 18 and has_license and has_car:
    print("You can drive!")
elif age >= 18 and has_license:
    print("You can drive, but you need a car")
elif age >= 18:
    print("You can get a license")
else:
    print("You're too young to drive")

# Example 5: Membership testing
vowels = "aeiou"
letter = "a"
if letter in vowels:
    print(f"{letter} is a vowel")
else:
    print(f"{letter} is a consonant")
````

### Float precision and comparisons

When working with floating-point numbers and if statements, you need to be aware of precision issues. Floats are stored in binary format, which can lead to small precision errors.

A famous example is that `0.1 + 0.2` does not exactly equal `0.3` in floating-point arithmetic. This means that direct equality comparisons with floats can sometimes fail unexpectedly.

Instead of exact equality checks, you should either:
1. Use the `round()` function to compare rounded values
2. Check if the difference between two floats is very small (tolerance-based comparison)

````python
# Run this cell to see floating-point precision issues with if statements

print("Direct comparison problems:")
result = 0.1 + 0.2
print(f"0.1 + 0.2 = {result}")
print(f"0.1 + 0.2 == 0.3: {result == 0.3}")
print(f"Actual result: {repr(result)}")

print("\nSolution 1: Using round() function")
if round(0.1 + 0.2, 1) == round(0.3, 1):
    print("✓ Using round(1) - they are equal!")
else:
    print("✗ Still not equal")

if round(0.1 + 0.2, 10) == round(0.3, 10):
    print("✓ Using round(10) - they are equal!")
else:
    print("✗ Still not equal")

print("\nSolution 2: Tolerance-based comparison")
tolerance = 0.0001
if abs(0.1 + 0.2 - 0.3) < tolerance:
    print("✓ Within tolerance - close enough!")
else:
    print("✗ Not within tolerance")

print("\nPractical example - price calculations:")
price = 0.1 + 0.1 + 0.1  # Three items at $0.10 each
expected_total = 0.3

if round(price, 2) == round(expected_total, 2):
    print(f"✓ Price check passed: ${round(price, 2)}")
else:
    print(f"✗ Price mismatch: ${price} vs ${expected_total}")
````

### Compact If-Else (Ternary Operator)

Python provides a **compact way to write simple if-else statements** using the **ternary operator** (also called conditional expression). This allows you to write a simple if-else statement in a single line.

The syntax is:
```python
value_if_true if condition else value_if_false
```

This is equivalent to:
```python
if condition:
    result = value_if_true
else:
    result = value_if_false
```

The compact form is useful for simple assignments and when you need to choose between two values based on a condition. However, don't overuse it - if the logic becomes complex, regular if-else statements are more readable.

````python
# Run this cell to see compact if-else constructions

# Example 1: Basic ternary operator
age = 20
status = "adult" if age >= 18 else "minor"
print(f"Age: {age}, Status: {status}")

# Example 2: Comparison with regular if-else
# Regular way:
temperature = 85
if temperature > 80:
    clothing = "shorts"
else:
    clothing = "pants"
print(f"Regular if-else: {clothing}")

# Compact way:
clothing_compact = "shorts" if temperature > 80 else "pants"
print(f"Compact form: {clothing_compact}")

# Example 3: Using in function calls
numbers = [1, 2, 3, 4, 5]
for num in numbers:
    print(f"{num} is {'even' if num % 2 == 0 else 'odd'}")

# Example 4: With mathematical operations
x = 10
y = 5
max_value = x if x > y else y
print(f"Maximum of {x} and {y} is {max_value}")

# Example 5: Handling division by zero
a = 10
b = 0
result = a / b if b != 0 else "Cannot divide by zero"
print(f"Division result: {result}")
````

### For Loops

**For loops** allow you to iterate through a sequence of items (like lists, strings, ranges) and execute code for each item. They are essential for automating repetitive tasks.

Basic syntax:
```python
for item in sequence:
    # code to execute for each item
```

Common patterns:
* `for item in list:` - iterate through list items
* `for i in range(n):` - iterate n times with index
* `for i, item in enumerate(list):` - get both index and item
* `for key, value in dict.items():` - iterate through dictionary
* `for char in string:` - iterate through string characters

For loops can also use `break` to exit early and `continue` to skip to the next iteration.

### The range() function

The `range()` function is commonly used with for loops to generate sequences of numbers:

* `range(n)` - generates numbers from 0 to n-1
* `range(start, stop)` - generates numbers from start to stop-1  
* `range(start, stop, step)` - generates numbers from start to stop-1, incrementing by step

Examples:
* `range(5)` → 0, 1, 2, 3, 4
* `range(2, 8)` → 2, 3, 4, 5, 6, 7
* `range(0, 10, 2)` → 0, 2, 4, 6, 8

````python
# Run this cell to see for loops in action

# Example 1: Iterating through a list
fruits = ["apple", "banana", "cherry"]
print("Fruits:")
for fruit in fruits:
    print(f"- {fruit}")

# Example 2: Using range()
print("\nCounting to 5:")
for i in range(1, 6):
    print(f"Count: {i}")

# Example 3: Iterating with index using enumerate()
colors = ["red", "green", "blue"]
print("\nColors with indices:")
for i, color in enumerate(colors):
    print(f"Index {i}: {color}")

# Example 4: Iterating through a string
word = "Python"
print(f"\nLetters in '{word}':")
for letter in word:
    print(letter)

# Example 5: Iterating through a dictionary
student_grades = {"Alice": 85, "Bob": 92, "Charlie": 78}
print("\nStudent grades:")
for name, grade in student_grades.items():
    print(f"{name}: {grade}")

# Example 6: Using break and continue
print("\nNumbers 1-10, skipping 5, stopping at 8:")
for num in range(1, 11):
    if num == 5:
        continue  # skip 5
    if num == 8:
        break     # stop at 8
    print(num)

# Example 7: Nested loops
print("\nMultiplication table (first 3 rows):")
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i} x {j} = {i*j}", end="  ")
    print()  # new line after each row
````

### While Loops

**While loops** repeatedly execute code as long as a condition remains true. Unlike for loops, which iterate over a known sequence, while loops continue until a condition becomes false.

Basic syntax:
```python
while condition:
    # code to execute while condition is True
    # usually includes code that will eventually make condition False
```

Key considerations:
* Make sure the condition will eventually become False to avoid infinite loops
* Often used when you don't know how many iterations you need
* Common to modify a counter or update a variable inside the loop
* Can use `break` to exit early and `continue` to skip to next iteration

While loops are useful for user input validation, game loops, and situations where you need to repeat until some condition is met.

````python
# Run this cell to see while loops in action

# Example 1: Basic while loop with counter
print("Countdown:")
count = 5
while count > 0:
    print(count)
    count -= 1  # Decrease count by 1
print("Blast off!")

# Example 2: Accumulating values
print("\nSum of numbers until we reach 50:")
total = 0
number = 1
while total < 50:
    total += number
    print(f"Added {number}, total now: {total}")
    number += 1

# Example 3: Finding a value in a list
print("\nSearching for 'banana' in fruits:")
fruits = ["apple", "orange", "banana", "grape"]
index = 0
found = False
while index < len(fruits) and not found:
    if fruits[index] == "banana":
        print(f"Found banana at index {index}")
        found = True
    else:
        print(f"Index {index}: {fruits[index]} - not banana")
    index += 1

# Example 4: Using break statement
print("\nNumbers until we hit a multiple of 7:")
num = 1
while True:  # Infinite loop...
    if num % 7 == 0:
        print(f"Hit multiple of 7: {num}")
        break  # ...but we break out here
    print(num)
    num += 1

# Example 5: Doubling until threshold
print("\nDoubling starting from 1:")
value = 1
while value < 100:
    print(value)
    value *= 2
print(f"Final value: {value}")

# Example 6: Input validation (simulated)
print("\nSimulated input validation:")
attempts = 0
max_attempts = 3
while attempts < max_attempts:
    # Simulate user input (normally would use input())
    simulated_input = "wrong" if attempts < 2 else "password"
    print(f"Attempt {attempts + 1}: Trying '{simulated_input}'")
    
    if simulated_input == "password":
        print("Access granted!")
        break
    else:
        print("Wrong password")
        attempts += 1
        
if attempts >= max_attempts:
    print("Too many attempts. Access denied.")
````

### Error Handling with Try/Except

**Error handling** allows your program to gracefully deal with unexpected situations instead of crashing. Python uses `try`/`except` blocks to catch and handle errors (called exceptions).

Basic syntax:
```python
try:
    # code that might cause an error
    risky_code()
except SpecificError:
    # handle specific error type
    print("Handle specific error")
except:
    # handle any other error
    print("Handle any error")
```

**Common exception types:**
- `ValueError` - invalid value for operation (e.g., `int("abc")`)
- `ZeroDivisionError` - division by zero
- `IndexError` - list index out of range  
- `KeyError` - dictionary key not found
- `FileNotFoundError` - file doesn't exist
- `TypeError` - wrong data type for operation

Error handling is essential for user input validation, file operations, and making robust programs.

````python
# Run this cell to see basic try/except in action

# Example 1: Handling division by zero
print("=== DIVISION BY ZERO ===")
try:
    result = 10 / 0
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

# Example 2: Handling invalid input conversion
print("\n=== INVALID INPUT CONVERSION ===")
user_input = "abc"  # Simulating user input
try:
    number = int(user_input)
    print(f"Number: {number}")
except ValueError:
    print(f"Error: '{user_input}' is not a valid number")

# Example 3: Handling list index errors
print("\n=== INDEX ERROR ===")
my_list = [1, 2, 3]
try:
    value = my_list[10]
    print(f"Value: {value}")
except IndexError:
    print(f"Error: Index out of range. List has {len(my_list)} items")

# Example 4: Handling dictionary key errors
print("\n=== KEY ERROR ===")
student_grades = {"Alice": 85, "Bob": 92}
try:
    grade = student_grades["Charlie"]
    print(f"Charlie's grade: {grade}")
except KeyError:
    print("Error: Student 'Charlie' not found in gradebook")

# Example 5: Multiple exception types
print("\n=== MULTIPLE EXCEPTIONS ===")
def safe_divide_and_access(numbers, index, divisor):
    try:
        result = numbers[index] / divisor
        return result
    except IndexError:
        return "Error: Index out of range"
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid data type"

numbers = [10, 20, 30]
print(f"Result 1: {safe_divide_and_access(numbers, 1, 2)}")  # Normal case
print(f"Result 2: {safe_divide_and_access(numbers, 5, 2)}")  # Index error
print(f"Result 3: {safe_divide_and_access(numbers, 1, 0)}")  # Division error
````

### Else and Finally Clauses

Try/except blocks can also include `else` and `finally` clauses for more sophisticated error handling:

**Complete syntax:**
```python
try:
    # code that might raise an exception
    risky_code()
except SpecificError:
    # handle the exception
    handle_error()
else:
    # runs only if NO exception occurred
    success_code()
finally:
    # ALWAYS runs, regardless of exceptions
    cleanup_code()
```

**When to use each:**
- **`else`**: Code that should only run if the try block succeeded (no exceptions)
- **`finally`**: Code that must always run (cleanup, closing files, etc.)

The `else` clause is useful for code that should only execute when the risky operation succeeds. The `finally` clause is essential for cleanup operations that must happen regardless of success or failure.

````python
# Run this cell to see else and finally clauses in action

# Example 1: Using else clause
print("=== ELSE CLAUSE ===")
def safe_division(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
        return None
    else:
        print(f"Division successful: {a} / {b} = {result}")
        return result

print("Test 1:")
safe_division(10, 2)  # Will use else clause
print("\nTest 2:")
safe_division(10, 0)  # Will use except clause

# Example 2: Using finally clause
print("\n=== FINALLY CLAUSE ===")
def process_data(data, index):
    print(f"Starting to process data at index {index}")
    try:
        value = data[index]
        processed = value * 2
        print(f"Processing successful: {value} * 2 = {processed}")
        return processed
    except IndexError:
        print(f"Error: Index {index} is out of range")
        return None
    finally:
        print("Cleanup: Process completed")

test_data = [1, 2, 3, 4, 5]
print("Test 1 (valid index):")
process_data(test_data, 2)
print("\nTest 2 (invalid index):")
process_data(test_data, 10)

# Example 3: Complete try/except/else/finally
print("\n=== COMPLETE STRUCTURE ===")
def read_and_process_number(text):
    print(f"Attempting to process: '{text}'")
    try:
        number = float(text)
        result = 100 / number
    except ValueError:
        print("Error: Invalid number format")
        return None
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
        return None
    else:
        print(f"Success: 100 / {number} = {result}")
        return result
    finally:
        print("Processing attempt completed")

print("Test with valid number:")
read_and_process_number("5.0")
print("\nTest with invalid format:")
read_and_process_number("abc")
print("\nTest with zero:")
read_and_process_number("0")

# Example 4: File handling simulation (showing finally for cleanup)
print("\n=== FILE HANDLING SIMULATION ===")
def simulate_file_processing(filename, will_fail=False):
    file_handle = None
    try:
        print(f"Opening file: {filename}")
        file_handle = f"Handle for {filename}"  # Simulated file handle
        
        if will_fail:
            raise Exception("Simulated file processing error")
        
        print("File processing successful")
        return "Success"
    except Exception as e:
        print(f"Error during file processing: {e}")
        return None
    else:
        print("No errors occurred during processing")
    finally:
        if file_handle:
            print(f"Closing file handle: {file_handle}")
        print("File operation completed")

print("Successful file processing:")
simulate_file_processing("data.txt", False)
print("\nFailed file processing:")
simulate_file_processing("bad_file.txt", True)
````

### List and Dictionary Comprehensions

**Comprehensions** provide a concise way to create lists and dictionaries using a compact syntax. They combine loops and conditional logic into a single expression.

**List Comprehensions:**
```python
new_list = [expression for item in iterable if condition]
```

**Dictionary Comprehensions:**
```python
new_dict = {key_expr: value_expr for item in iterable if condition}
```

**Set Comprehensions:**
```python
new_set = {expression for item in iterable if condition}
```

Comprehensions are often more readable and faster than equivalent for loops. They're particularly useful for transforming data, filtering collections, and creating new data structures from existing ones.

The `if condition` part is optional and can be used to filter which items to include.

````python
# Run this cell to see comprehensions in action

# List Comprehensions
print("=== LIST COMPREHENSIONS ===")

# Example 1: Square numbers
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]
print(f"Numbers: {numbers}")
print(f"Squares: {squares}")

# Equivalent with for loop:
squares_loop = []
for x in numbers:
    squares_loop.append(x**2)
print(f"Squares (for loop): {squares_loop}")

# Example 2: Filter even numbers and double them
evens_doubled = [x * 2 for x in range(10) if x % 2 == 0]
print(f"Even numbers doubled: {evens_doubled}")

# Example 3: String manipulation
words = ["hello", "world", "python", "programming"]
uppercase_long = [word.upper() for word in words if len(word) > 5]
print(f"Long words in uppercase: {uppercase_long}")

# Dictionary Comprehensions
print("\n=== DICTIONARY COMPREHENSIONS ===")

# Example 4: Create a dictionary of squares
square_dict = {x: x**2 for x in range(1, 6)}
print(f"Square dictionary: {square_dict}")

# Example 5: Transform existing dictionary
prices = {"apple": 1.20, "banana": 0.50, "orange": 0.80}
expensive_items = {item: price for item, price in prices.items() if price > 0.75}
print(f"Expensive items: {expensive_items}")

# Example 6: Create dictionary from two lists
fruits = ["apple", "banana", "cherry"]
colors = ["red", "yellow", "red"]
fruit_colors = {fruit: color for fruit, color in zip(fruits, colors)}
print(f"Fruit colors: {fruit_colors}")

# Set Comprehensions
print("\n=== SET COMPREHENSIONS ===")

# Example 7: Unique word lengths
words = ["hello", "world", "hello", "python", "world"]
unique_lengths = {len(word) for word in words}
print(f"Words: {words}")
print(f"Unique word lengths: {unique_lengths}")

# Nested Comprehensions
print("\n=== NESTED COMPREHENSIONS ===")

# Example 8: Flatten a 2D list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
print(f"Matrix: {matrix}")
print(f"Flattened: {flattened}")

# Example 9: Comprehension vs traditional approach
print("\n=== PERFORMANCE COMPARISON ===")
import time

# Traditional approach
start_time = time.time()
traditional = []
for i in range(1000):
    if i % 2 == 0:
        traditional.append(i * 2)
traditional_time = time.time() - start_time

# Comprehension approach
start_time = time.time()
comprehension = [i * 2 for i in range(1000) if i % 2 == 0]
comprehension_time = time.time() - start_time

print(f"Traditional loop time: {traditional_time:.6f} seconds")
print(f"Comprehension time: {comprehension_time:.6f} seconds")
print(f"Results are equal: {traditional == comprehension}")
````
