### Introduction to Functions

**Functions** are reusable blocks of code that perform specific tasks. They are one of the most important concepts in programming, allowing you to:

- **Organize code** into logical, manageable pieces
- **Avoid repetition** by writing code once and using it multiple times
- **Make code more readable** by giving meaningful names to operations
- **Debug more easily** by isolating functionality
- **Build complex programs** by combining simple functions

**How Functions Work:**
1. **Definition**: You define a function using the `def` keyword
2. **Parameters**: Functions can accept input values (parameters)
3. **Body**: The function contains code that processes the inputs
4. **Return**: Functions can return results back to the caller
5. **Calling**: You execute a function by calling it with arguments

Python provides many built-in functions (like `print()`, `len()`, `max()`), and you can create your own custom functions to solve specific problems.

### Built-in Functions

Python comes with many **built-in functions** that you can use immediately without importing anything. These functions handle common tasks and operations, making your code more efficient and readable.

**Mathematical Functions:**
- **`abs(x)`** - returns absolute value
- **`max(iterable)`** - finds maximum value
- **`min(iterable)`** - finds minimum value  
- **`sum(iterable)`** - calculates sum of numbers
- **`round(number, digits)`** - rounds to specified decimal places
- **`pow(base, exponent)`** - calculates base raised to power
- **`divmod(a, b)`** - returns quotient and remainder

**Sequence Functions:**
- **`len(sequence)`** - returns length of sequence
- **`sorted(iterable)`** - returns sorted list
- **`reversed(sequence)`** - returns reversed iterator
- **`enumerate(iterable)`** - adds index numbers to items
- **`zip(*iterables)`** - combines multiple sequences

**Type and Conversion Functions:**
- **`type(object)`** - returns type of object
- **`str(object)`**, **`int(object)`**, **`float(object)`**, **`bool(object)`** - convert types
- **`isinstance(object, type)`** - checks if object is of specified type

**Logic Functions:**
- **`any(iterable)`** - True if any element is True
- **`all(iterable)`** - True if all elements are True

These built-in functions are optimized and should be used whenever possible instead of writing your own implementations.

````python
# Run this cell to see built-in functions in action

# Mathematical Functions
print("=== MATHEMATICAL FUNCTIONS ===")
numbers = [-5, 3, -1, 8, 2]
print(f"Numbers: {numbers}")
print(f"abs(-5): {abs(-5)}")
print(f"max(numbers): {max(numbers)}")
print(f"min(numbers): {min(numbers)}")
print(f"sum(numbers): {sum(numbers)}")
print(f"round(3.14159, 2): {round(3.14159, 2)}")
print(f"pow(2, 5): {pow(2, 5)}")
print(f"divmod(17, 5): {divmod(17, 5)}")  # Returns (quotient, remainder)

# Sequence Functions
print("\n=== SEQUENCE FUNCTIONS ===")
fruits = ["banana", "apple", "cherry", "date"]
print(f"Fruits: {fruits}")
print(f"len(fruits): {len(fruits)}")
print(f"sorted(fruits): {sorted(fruits)}")
print(f"list(reversed(fruits)): {list(reversed(fruits))}")

# Enumerate adds index numbers
print(f"\nUsing enumerate:")
for index, fruit in enumerate(fruits):
    print(f"  {index}: {fruit}")

# Zip combines sequences
scores = [85, 92, 78, 95]
print(f"\nScores: {scores}")
print(f"Zip fruits with scores:")
for fruit, score in zip(fruits, scores):
    print(f"  {fruit}: {score}")

# Type and Conversion Functions
print("\n=== TYPE AND CONVERSION FUNCTIONS ===")
value = 42
print(f"Value: {value}")
print(f"type(value): {type(value)}")
print(f"str(value): '{str(value)}'")
print(f"float(value): {float(value)}")
print(f"bool(value): {bool(value)}")
print(f"isinstance(value, int): {isinstance(value, int)}")
print(f"isinstance(value, str): {isinstance(value, str)}")

# Converting strings to numbers
text_numbers = ["10", "20", "30"]
print(f"\nText numbers: {text_numbers}")
numeric = [int(x) for x in text_numbers]
print(f"Converted to integers: {numeric}")

# Logic Functions
print("\n=== LOGIC FUNCTIONS ===")
bool_list1 = [True, False, True]
bool_list2 = [True, True, True]
bool_list3 = [False, False, False]

print(f"List 1: {bool_list1}")
print(f"any(bool_list1): {any(bool_list1)}")
print(f"all(bool_list1): {all(bool_list1)}")

print(f"\nList 2: {bool_list2}")
print(f"any(bool_list2): {any(bool_list2)}")
print(f"all(bool_list2): {all(bool_list2)}")

print(f"\nList 3: {bool_list3}")
print(f"any(bool_list3): {any(bool_list3)}")
print(f"all(bool_list3): {all(bool_list3)}")

# Practical examples with numbers
test_numbers = [2, 4, 6, 8, 10]
print(f"\n=== PRACTICAL EXAMPLES ===")
print(f"Numbers: {test_numbers}")
print(f"Are all numbers even? {all(num % 2 == 0 for num in test_numbers)}")
print(f"Any numbers > 5? {any(num > 5 for num in test_numbers)}")

# Working with strings
text = "Hello World"
print(f"\nText: '{text}'")
print(f"Length: {len(text)}")
print(f"Max character: '{max(text)}'")  # Lexicographically largest
print(f"Min character: '{min(text)}'")  # Lexicographically smallest

# Range function (often used with other built-ins)
print(f"\n=== RANGE WITH OTHER FUNCTIONS ===")
range_nums = list(range(1, 11))
print(f"Range 1-10: {range_nums}")
print(f"Sum of 1-10: {sum(range(1, 11))}")
print(f"Max of range: {max(range(1, 11))}")

# Multiple operations combined
grades = [88, 92, 76, 95, 83]
print(f"\nGrades: {grades}")
print(f"Number of grades: {len(grades)}")
print(f"Average grade: {sum(grades) / len(grades):.1f}")
print(f"Highest grade: {max(grades)}")
print(f"Lowest grade: {min(grades)}")
print(f"All grades passing (>=70)? {all(grade >= 70 for grade in grades)}")
print(f"Any grade is A (>=90)? {any(grade >= 90 for grade in grades)}")
````

### Creating Your Own Functions

You can create your own **custom functions** to encapsulate code that you want to reuse or organize. Creating functions helps make your code more modular, readable, and maintainable.

**Basic Function Syntax:**
```python
def function_name(parameter1, parameter2):
    """Optional docstring explaining what the function does"""
    # Function body
    result = parameter1 + parameter2
    return result  # Optional return statement
```

**Key Components:**
- **`def`** keyword starts the function definition
- **Function name** should be descriptive and follow Python naming conventions (lowercase with underscores)
- **Parameters** are input values the function accepts (optional)
- **Docstring** provides documentation (optional but recommended)
- **Function body** contains the code to execute (must be indented)
- **`return`** statement sends a value back to the caller (optional)

**Function Types:**
- **Functions that return values** - perform calculations and return results
- **Functions that perform actions** - do something but don't return a value
- **Functions with parameters** - accept input values
- **Functions without parameters** - no input needed

**Function Naming Conventions:**
- Use descriptive names that explain what the function does
- Use lowercase letters with underscores: `calculate_area`, `is_valid`
- Avoid single letters except for very simple cases

````python
# Run this cell to see custom functions in action

# Example 1: Simple function with parameters and return value
def add_numbers(a, b):
    """Add two numbers and return the result"""
    return a + b

result = add_numbers(5, 3)
print(f"add_numbers(5, 3) = {result}")

# Example 2: Function without return (performs action)
def greet_person(name):
    """Print a greeting message"""
    print(f"Hello, {name}! Welcome to Python!")

greet_person("Alice")

# Example 3: Function with no parameters
def get_current_status():
    """Return a status message"""
    return "System is running normally"

status = get_current_status()
print(f"Status: {status}")

# Example 4: Function with multiple operations and docstring
def calculate_circle_area(radius):
    """
    Calculate the area of a circle given its radius.
    
    Parameters:
    radius (float): The radius of the circle
    
    Returns:
    float: The area of the circle
    """
    pi = 3.14159
    area = pi * radius * radius
    return area

circle_area = calculate_circle_area(5)
print(f"Area of circle with radius 5: {circle_area:.2f}")

# Example 5: Function with conditional logic
def classify_temperature(temp):
    """Classify temperature as freezing, cold, moderate, or hot"""
    if temp < 32:
        return "freezing"
    elif temp < 70:
        return "cold"
    elif temp < 85:
        return "moderate"
    else:
        return "hot"

temp_classification = classify_temperature(75)
print(f"75°F is classified as: {temp_classification}")

# Example 6: Function that processes a list
def count_positive_numbers(numbers):
    """Count how many positive numbers are in a list"""
    count = 0
    for num in numbers:
        if num > 0:
            count += 1
    return count

numbers_list = [-2, 5, -1, 8, 0, 3]
positive_count = count_positive_numbers(numbers_list)
print(f"Positive numbers in {numbers_list}: {positive_count}")

# Example 7: Function with string manipulation
def format_name(first, last):
    """Format a person's name in 'Last, First' format"""
    return f"{last}, {first}"

formatted = format_name("John", "Smith")
print(f"Formatted name: {formatted}")

# Example 8: Function that uses built-in functions
def get_list_stats(numbers):
    """Get basic statistics about a list of numbers"""
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    maximum = max(numbers)
    minimum = min(numbers)
    
    return f"Count: {count}, Sum: {total}, Avg: {average:.1f}, Max: {maximum}, Min: {minimum}"

stats = get_list_stats([10, 25, 30, 15, 20])
print(f"List statistics: {stats}")

# Example 9: Function with validation
def safe_divide(a, b):
    """Safely divide two numbers, handling division by zero"""
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b

print(f"safe_divide(10, 2) = {safe_divide(10, 2)}")
print(f"safe_divide(10, 0) = {safe_divide(10, 0)}")

# Accessing function documentation
print(f"\nDocumentation for calculate_circle_area:")
print(calculate_circle_area.__doc__)
````

### Variable Scope

**Variable scope** determines where in your code a variable can be accessed. Understanding scope is crucial when working with functions.

**Local Scope:**
- Variables created inside a function are **local** to that function
- Local variables can only be accessed within the function where they're defined
- Each function call creates its own set of local variables

**Global Scope:**
- Variables created outside functions are **global**
- Global variables can be accessed from anywhere in your code
- Functions can read global variables but creating variables with the same name creates local versions

**Key Rules:**
1. **Function parameters are local variables**
2. **Variables assigned inside functions are local** (unless declared global)
3. **Local variables shadow global variables** with the same name
4. **Functions can read but not modify global variables** (without `global` keyword)

**Best Practice:** Use local variables whenever possible to avoid unexpected interactions between different parts of your code.

````python
# Run this cell to see variable scope in action

# Global variable
global_counter = 100
company_name = "TechCorp"

print("=== GLOBAL VARIABLES ===")
print(f"Global counter: {global_counter}")
print(f"Company name: {company_name}")

# Example 1: Function using local variables
def calculate_bonus(salary, percentage):
    """Calculate bonus using local variables"""
    bonus = salary * (percentage / 100)  # bonus is local
    tax_rate = 0.25  # tax_rate is local
    final_bonus = bonus * (1 - tax_rate)  # final_bonus is local
    return final_bonus

employee_bonus = calculate_bonus(50000, 10)
print(f"\nEmployee bonus: ${employee_bonus:.2f}")

# Trying to access local variables outside the function would cause an error:
# print(bonus)  # This would cause: NameError: name 'bonus' is not defined

# Example 2: Function reading global variables
def display_company_info():
    """Function that reads global variables"""
    print(f"Working at {company_name}")  # Reads global variable
    print(f"Current counter value: {global_counter}")  # Reads global variable

print("\n=== READING GLOBAL VARIABLES ===")
display_company_info()

# Example 3: Local variable shadowing global variable
def test_shadowing():
    """Demonstrate local variable shadowing global variable"""
    global_counter = 50  # This creates a LOCAL variable with the same name
    print(f"Inside function - counter: {global_counter}")  # Uses local version

print("\n=== VARIABLE SHADOWING ===")
print(f"Before function call - global counter: {global_counter}")
test_shadowing()
print(f"After function call - global counter: {global_counter}")  # Global unchanged

# Example 4: Function parameters are local variables
def process_data(data, multiplier):
    """Function parameters are local variables"""
    data = data * multiplier  # Modifying the local parameter
    result = data + 10  # result is also local
    print(f"Inside function - data: {data}, result: {result}")
    return result

original_data = 5
original_multiplier = 3
print(f"\n=== FUNCTION PARAMETERS ===")
print(f"Before function - data: {original_data}, multiplier: {original_multiplier}")
returned_value = process_data(original_data, original_multiplier)
print(f"After function - data: {original_data}, multiplier: {original_multiplier}")  # Unchanged
print(f"Returned value: {returned_value}")

# Example 5: Multiple functions with their own local scope
def function_a():
    """First function with local variables"""
    x = 10
    y = 20
    print(f"Function A - x: {x}, y: {y}")
    return x + y

def function_b():
    """Second function with different local variables"""
    x = 100  # Different local variable with same name
    z = 300
    print(f"Function B - x: {x}, z: {z}")
    return x * 2

print(f"\n=== INDEPENDENT LOCAL SCOPES ===")
result_a = function_a()
result_b = function_b()
print(f"Result A: {result_a}, Result B: {result_b}")

# Example 6: Nested function calls
def outer_function(a):
    """Outer function that calls another function"""
    b = a * 2  # Local to outer_function
    print(f"Outer function - a: {a}, b: {b}")
    
    def inner_calculation(x):
        """Inner function with its own local scope"""
        y = x + 5  # Local to inner_calculation
        print(f"Inner function - x: {x}, y: {y}")
        return y
    
    result = inner_calculation(b)
    print(f"Outer function - result from inner: {result}")
    return result + b

print(f"\n=== NESTED FUNCTIONS ===")
final_result = outer_function(3)
print(f"Final result: {final_result}")

# Example 7: Common scope mistake
counter = 0

def increment_counter():
    """This function can READ the global counter but can't modify it directly"""
    print(f"Current counter value: {counter}")
    # counter = counter + 1  # This would cause UnboundLocalError
    # To modify global variables, you need the 'global' keyword (advanced topic)

print(f"\n=== READING GLOBAL VARIABLES ===")
increment_counter()
print(f"Global counter remains: {counter}")

# Example 8: Function scope isolation
def test_isolation():
    """Test that functions don't interfere with each other"""
    temp_value = 999
    return temp_value

temp_value = 1  # Global variable
print(f"\n=== SCOPE ISOLATION ===")
print(f"Global temp_value: {temp_value}")
function_result = test_isolation()
print(f"Function returned: {function_result}")
print(f"Global temp_value after function: {temp_value}")  # Unchanged
````

### Functions Returning Multiple Values

Python functions can **return multiple values** at once, which is useful when you need to return several related pieces of information. This is accomplished using **tuples**.

**Returning Multiple Values:**
```python
def function_name():
    value1 = "first"
    value2 = "second"
    return value1, value2  # Returns a tuple
```

**Unpacking Return Values:**
```python
# Method 1: Unpack into separate variables
first, second = function_name()

# Method 2: Receive as a tuple
result = function_name()  # result is a tuple
```

**Common Use Cases:**
- Returning coordinates (x, y)
- Returning statistics (mean, median, mode)
- Returning min and max values
- Returning multiple calculations
- Returning status and result
- Returning processed data with metadata

This feature makes functions more versatile and allows you to avoid calling multiple functions when you need related information.

````python
# Run this cell to see functions returning multiple values

# Example 1: Basic multiple return values
def get_name_parts():
    """Return first and last name"""
    return "John", "Smith"

first, last = get_name_parts()
print(f"First name: {first}, Last name: {last}")

# You can also receive as a tuple
full_name = get_name_parts()
print(f"Full name tuple: {full_name}")
print(f"Accessing tuple elements: {full_name[0]} {full_name[1]}")

# Example 2: Mathematical calculations
def rectangle_properties(length, width):
    """Calculate area and perimeter of a rectangle"""
    area = length * width
    perimeter = 2 * (length + width)
    return area, perimeter

area, perimeter = rectangle_properties(5, 3)
print(f"\nRectangle 5x3 - Area: {area}, Perimeter: {perimeter}")

# Example 3: Statistics from a list using built-in functions
def get_statistics(numbers):
    """Return min, max, and average of a list"""
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    return minimum, maximum, average

data = [10, 5, 8, 12, 3, 15]
min_val, max_val, avg_val = get_statistics(data)
print(f"\nData: {data}")
print(f"Min: {min_val}, Max: {max_val}, Average: {avg_val:.2f}")

# Example 4: Coordinate operations
def translate_point(x, y, dx, dy):
    """Move a point by given offsets"""
    new_x = x + dx
    new_y = y + dy
    return new_x, new_y

original_x, original_y = 10, 20
new_x, new_y = translate_point(original_x, original_y, 5, -3)
print(f"\nPoint moved from ({original_x}, {original_y}) to ({new_x}, {new_y})")

# Example 5: String analysis
def analyze_string(text):
    """Return length, number of words, and uppercase version"""
    length = len(text)
    word_count = len(text.split())
    uppercase = text.upper()
    return length, word_count, uppercase

text = "Hello Python World"
length, words, upper = analyze_string(text)
print(f"\nText: '{text}'")
print(f"Length: {length}, Words: {words}, Uppercase: '{upper}'")

# Example 6: Processing with status
def process_grade(score):
    """Process a grade and return the letter grade and pass status"""
    if score >= 90:
        letter = "A"
    elif score >= 80:
        letter = "B"
    elif score >= 70:
        letter = "C"
    elif score >= 60:
        letter = "D"
    else:
        letter = "F"
    
    passed = score >= 60
    return letter, passed

grade_letter, is_passing = process_grade(85)
print(f"\nScore 85: Letter grade {grade_letter}, Passing: {is_passing}")

# Example 7: Built-in divmod function (returns multiple values)
quotient, remainder = divmod(17, 5)
print(f"\n17 divided by 5: quotient = {quotient}, remainder = {remainder}")

# Example 8: Function returning different numbers of values based on condition
def get_circle_properties(radius, include_diameter=False):
    """Get circle properties, optionally including diameter"""
    area = 3.14159 * radius * radius
    circumference = 2 * 3.14159 * radius
    
    if include_diameter:
        diameter = 2 * radius
        return area, circumference, diameter
    else:
        return area, circumference

# Different unpacking based on what's returned
area1, circumference1 = get_circle_properties(5)
print(f"\nCircle radius 5: Area={area1:.1f}, Circumference={circumference1:.1f}")

area2, circumference2, diameter2 = get_circle_properties(5, include_diameter=True)
print(f"With diameter: Area={area2:.1f}, Circumference={circumference2:.1f}, Diameter={diameter2}")

# Example 9: Practical data processing
def process_sales_data(sales):
    """Process sales data and return summary statistics"""
    total_sales = sum(sales)
    num_transactions = len(sales)
    average_sale = total_sales / num_transactions
    best_sale = max(sales)
    worst_sale = min(sales)
    
    return total_sales, num_transactions, average_sale, best_sale, worst_sale

sales_data = [150, 200, 75, 300, 125, 180]
total, count, avg, best, worst = process_sales_data(sales_data)
print(f"\nSales Analysis:")
print(f"Total: ${total}, Transactions: {count}, Average: ${avg:.2f}")
print(f"Best Sale: ${best}, Worst Sale: ${worst}")
````

### Functions with Default Parameters and Named Arguments

Python functions support **named arguments** (also called keyword arguments) and **default parameters**, which make function calls more readable and flexible.

**Default Parameters:**
```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"
```

**Calling with Named Arguments:**
```python
# Positional arguments
result = greet("Alice", "Hi")

# Named arguments
result = greet(name="Alice", greeting="Hi")
result = greet(greeting="Hi", name="Alice")  # Order doesn't matter

# Mix of positional and named
result = greet("Alice", greeting="Hi")
```

**Benefits:**
- **Clarity** - makes function calls self-documenting
- **Flexibility** - can call with arguments in any order
- **Default values** - optional parameters with sensible defaults
- **Extensibility** - easy to add new optional parameters

**Best Practices:**
- Use default values for optional parameters
- Put required parameters first, optional ones last
- Use descriptive parameter names
- Don't use mutable objects (lists, dicts) as default values

````python
# Run this cell to see named arguments and default parameters

# Example 1: Function with default parameters
def introduce_person(name, age, city="Unknown"):
    """Introduce a person with optional city"""
    return f"Hi, I'm {name}, {age} years old, from {city}."

# Different ways to call the function
print("=== DEFAULT PARAMETERS ===")
print(introduce_person("Alice", 25))  # Uses default city
print(introduce_person("Bob", 30, "New York"))  # Provides city
print(introduce_person("Charlie", 35, city="Boston"))  # Named argument for city

# Example 2: Function with multiple default parameters
def format_price(amount, currency="USD", include_symbol=True):
    """Format a price with currency options"""
    symbol_map = {"USD": "$", "EUR": "€", "GBP": "£"}
    
    if include_symbol and currency in symbol_map:
        return f"{symbol_map[currency]}{amount:.2f}"
    else:
        return f"{amount:.2f} {currency}"

print("\n=== MULTIPLE DEFAULT PARAMETERS ===")
print(format_price(29.99))  # All defaults
print(format_price(29.99, "EUR"))  # Custom currency
print(format_price(29.99, currency="GBP"))  # Named argument
print(format_price(29.99, include_symbol=False))  # Override symbol
print(format_price(29.99, currency="EUR", include_symbol=False))  # Multiple named

# Example 3: Named arguments in any order
def create_user_profile(username, email, first_name, last_name, active=True):
    """Create a user profile"""
    status = "Active" if active else "Inactive"
    return {
        "username": username,
        "email": email,
        "full_name": f"{first_name} {last_name}",
        "status": status
    }

print("\n=== NAMED ARGUMENTS IN ANY ORDER ===")
# Positional order
profile1 = create_user_profile("john_doe", "john@email.com", "John", "Doe")
print(f"Profile 1: {profile1}")

# Named arguments (different order)
profile2 = create_user_profile(
    last_name="Smith",
    first_name="Jane", 
    email="jane@email.com",
    username="jane_smith",
    active=False
)
print(f"Profile 2: {profile2}")

# Example 4: Mathematical function with options
def calculate_compound_interest(principal, rate, time, compound_frequency=1):
    """Calculate compound interest with optional compounding frequency"""
    amount = principal * (1 + rate/compound_frequency) ** (compound_frequency * time)
    interest = amount - principal
    return amount, interest

print("\n=== COMPOUND INTEREST CALCULATIONS ===")
# Simple annual compounding
amount1, interest1 = calculate_compound_interest(1000, 0.05, 2)
print(f"$1000 at 5% for 2 years (annual): Amount=${amount1:.2f}, Interest=${interest1:.2f}")

# Monthly compounding
amount2, interest2 = calculate_compound_interest(1000, 0.05, 2, compound_frequency=12)
print(f"$1000 at 5% for 2 years (monthly): Amount=${amount2:.2f}, Interest=${interest2:.2f}")

# Example 5: Text processing with options
def process_text(text, uppercase=False, remove_spaces=False, add_prefix=""):
    """Process text with various options"""
    result = text
    
    if uppercase:
        result = result.upper()
    
    if remove_spaces:
        result = result.replace(" ", "")
    
    if add_prefix:
        result = add_prefix + result
    
    return result

print("\n=== TEXT PROCESSING EXAMPLES ===")
original = "hello world"
print(f"Original: '{original}'")
print(f"Uppercase: '{process_text(original, uppercase=True)}'")
print(f"No spaces: '{process_text(original, remove_spaces=True)}'")
print(f"With prefix: '{process_text(original, add_prefix=\">>> \")}'")
print(f"All options: '{process_text(original, uppercase=True, remove_spaces=True, add_prefix=\"DATA: \")}'")

# Example 6: Data analysis function with many defaults
def analyze_numbers(numbers, show_details=True, sort_data=False, round_to=2):
    """Analyze a list of numbers with various display options"""
    if sort_data:
        numbers = sorted(numbers)
    
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    minimum = min(numbers)
    maximum = max(numbers)
    
    if round_to is not None:
        average = round(average, round_to)
    
    if show_details:
        return {
            "data": numbers,
            "count": count,
            "sum": total,
            "average": average,
            "min": minimum,
            "max": maximum
        }
    else:
        return {"average": average, "count": count}

print("\n=== DATA ANALYSIS EXAMPLES ===")
test_data = [15, 23, 8, 42, 16, 4]

# Default behavior
result1 = analyze_numbers(test_data)
print("Default analysis:", result1)

# Sorted with limited details
result2 = analyze_numbers(test_data, show_details=False, sort_data=True)
print("Limited analysis (sorted):", result2)

# All options
result3 = analyze_numbers(test_data, show_details=True, sort_data=True, round_to=1)
print("Full analysis (sorted, 1 decimal):", result3)
````
