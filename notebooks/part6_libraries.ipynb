### Example

There is no built-in square root function in python.  However, there is a `math` library that contains a square root function.  Run the following cell to see one way the square root function can be imported into a session.

````python
from math import sqrt

w = sqrt(9)
print(w)
````

### Aliases 

We would not want to do so in this example, but we could also change the name when we import it using `as`.  This is usually done to abbreviate long names.

````python
from math import sqrt as another_name

x = another_name(9)
print(x)
````

### Importing Libraries 

Some python libraries contain many useful things, so we will import the entire library rather than importing objects one at a time.  Here is an example.

````python
import math

y = math.sqrt(9)
print(y)
````

Pay attention to the syntax here.  We prefaced `sqrt` with `math.` because `sqrt` is a function in the `math` library, and we imported the library in this example but not the function explicitly.

### Aliases for Libraries

We can use `as` when importing entire libraries to abbreviate the library name.  Again, we would not do it here, because `math` is a short name, but we will illustrate using `math` anyway.

````python
import math as another_name

z = another_name.sqrt(9)
print(z)
````

### Alias Conventions 

We will use python libraries throughout the remainder of the workshop.  There are conventions regarding the aliases used when importing these libraries.  It is not necessary to follow the conventions, but everyone does.  Here are the conventions for standard libraries.

- pandas is the main library for data handling
- statsmodels is for statistics, including regression
- matplotlib and seaborn are for plotting
- numpy is numerical python (calculations on arrays)

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

These conventions are understood by AI.  If you ask the Gemini chatbot to explain `np.array` it will "know" that you are asking about the `array` function from `numpy`.

### Installing Libraries with pip

Python comes with many built-in libraries (like `math`, `datetime`, ...), but you'll often need to install additional **third-party libraries** for specialized tasks.

Google Colab has already installed many standard libraries, including those listed above, but if we need to install something else, we do it with the `pip` command (Python's package manager) by adding an exclamation mark (`!`) before the command, as in this example.

```python
!pip install yfinance
```

We can ensure we get the latest version of the library with

```python
!pip install --upgrade yfinance
```

and we can install a specific version with

```python
!pip install yfinance==0.2.50
```

- Installations are **temporary** - you'll need to reinstall if you start a new session
- You can check installed packages with `!pip list`

````python
# Run this cell to see different import patterns in action

print("=== 1. IMPORT ENTIRE MODULE ===")
import math
import datetime

# Need to use module name as prefix
circle_area = math.pi * (5 ** 2)
current_time = datetime.datetime.now()
print(f"Circle area: {circle_area:.2f}")
print(f"Current time: {current_time}")

print("\n=== 2. IMPORT WITH ALIAS ===")
import math as m
import datetime as dt

# Use shorter alias
circle_area2 = m.pi * (3 ** 2)
tomorrow = dt.datetime.now() + dt.timedelta(days=1)
print(f"Circle area: {circle_area2:.2f}")
print(f"Tomorrow: {tomorrow.date()}")

print("\n=== 3. IMPORT SPECIFIC FUNCTIONS ===")
from math import sqrt, pi, cos
from datetime import date, timedelta

# Use functions directly without prefix
hypotenuse = sqrt(3**2 + 4**2)
today = date.today()
next_week = today + timedelta(weeks=1)
print(f"Hypotenuse: {hypotenuse}")
print(f"Today: {today}")
print(f"Next week: {next_week}")
````
