### Overview

A python library is a collection of python code.  When you import a python library, you gain access to the additional functionality provided by the code.

### Example

There is no built-in square root function in python.  However, there is a `math` library that contains a square root function.  Run the following cell to see one way the square root function can be imported into a session.

````python
from math import sqrt

w = sqrt(9)
print(w)
````

### Aliases 

We would not want to do so in this example, but we could also change the name when we import it using `as`.  This is usually done to abbreviate long names.

````python
from math import sqrt as alias_for_sqrt

x = alias_for_sqrt(9)
print(x)
````

### Importing Libraries 

Some python libraries contain many useful things, so we will import the entire library rather than importing objects one at a time.  Here is an example.

When we import an entire library, we must use the library name as a prefix afterwards to access the library objects.

````python
import math

y = math.sqrt(9)
print(y)
````

We can use `as` when importing entire libraries to abbreviate the library name.  Again, we would not do it here, because `math` is a short name, but we will illustrate using `math` anyway.

### Common Import Patterns and When to Use Each

Python offers several ways to import libraries and their components. Each pattern has different use cases and trade-offs:

#### **1. Import Entire Module**
```python
import math
result = math.sqrt(16)
```
**When to use:** When you need many functions from a library, or when you want to avoid name conflicts.
**Pros:** Clear namespace, no conflicts, see where functions come from
**Cons:** More verbose, requires typing module name each time

#### **2. Import with Alias**  
```python
import pandas as pd
import numpy as np
```
**When to use:** For libraries with long names or standard conventions (pandas→pd, numpy→np)
**Pros:** Shorter typing, follows community conventions, clear namespace
**Cons:** Need to remember aliases

#### **3. Import Specific Functions**
```python
from math import sqrt, pi
result = sqrt(16) * pi
```
**When to use:** When you only need a few specific functions and want clean code
**Pros:** Direct access, cleaner code, no prefix needed
**Cons:** Potential name conflicts, unclear where functions come from in large code

#### **4. Import All (NOT Recommended)**
```python
from math import *  # DON'T DO THIS
```
**When to use:** Almost never - only in interactive sessions for quick testing
**Pros:** Access to everything without prefixes
**Cons:** Name conflicts, unclear code, imports unnecessary items

#### **5. Import Submodules**
```python
import matplotlib.pyplot as plt
from collections import Counter
```
**When to use:** When you need specific submodules of larger packages
**Pros:** Access to specialized functionality
**Cons:** Need to understand package structure

````python
# Run this cell to see different import patterns in action

print("=== 1. IMPORT ENTIRE MODULE ===")
import math
import datetime

# Need to use module name as prefix
circle_area = math.pi * (5 ** 2)
current_time = datetime.datetime.now()
print(f"Circle area: {circle_area:.2f}")
print(f"Current time: {current_time}")

print("\n=== 2. IMPORT WITH ALIAS ===")
import math as m
import datetime as dt

# Use shorter alias
circle_area2 = m.pi * (3 ** 2)
tomorrow = dt.datetime.now() + dt.timedelta(days=1)
print(f"Circle area: {circle_area2:.2f}")
print(f"Tomorrow: {tomorrow.date()}")

print("\n=== 3. IMPORT SPECIFIC FUNCTIONS ===")
from math import sqrt, pi, cos
from datetime import date, timedelta

# Use functions directly without prefix
hypotenuse = sqrt(3**2 + 4**2)
today = date.today()
next_week = today + timedelta(weeks=1)
print(f"Hypotenuse: {hypotenuse}")
print(f"Today: {today}")
print(f"Next week: {next_week}")

print("\n=== 4. IMPORT ALL (DEMONSTRATION ONLY) ===")
# This is generally NOT recommended, but here's how it works:
from random import *  # Imports everything from random
random_number = randint(1, 10)  # Can use randint directly
print(f"Random number: {random_number}")
# Note: This makes it unclear where 'randint' comes from!

print("\n=== 5. IMPORT SUBMODULES ===")
from collections import Counter, defaultdict
from urllib.parse import urlparse

# Using specialized submodules
word_counts = Counter(['apple', 'banana', 'apple', 'cherry', 'apple'])
parsed_url = urlparse('https://example.com/path?query=value')
print(f"Word counts: {word_counts}")
print(f"Parsed URL domain: {parsed_url.netloc}")

print("\n=== BEST PRACTICES EXAMPLES ===")

# Standard data science imports (conventional aliases)
try:
    import pandas as pd
    import numpy as np
    
    # Creating sample data
    data = pd.DataFrame({
        'values': [1, 2, 3, 4, 5],
        'squares': [1, 4, 9, 16, 25]
    })
    mean_value = np.mean(data['values'])
    print(f"Sample data:\n{data}")
    print(f"Mean value: {mean_value}")
    
except ImportError:
    print("pandas and/or numpy not available (not pre-installed)")

# Mixed import patterns in practice
from math import factorial
import random

def calculate_permutations(n, r):
    """Calculate permutations using mixed imports"""
    return factorial(n) // factorial(n - r)

# Generate random sample for demonstration
sample_size = random.randint(5, 10)
subset_size = random.randint(2, 4)
perms = calculate_permutations(sample_size, subset_size)
print(f"\nPermutations of {subset_size} items from {sample_size}: {perms}")

print("\n=== NAMESPACE COMPARISON ===")
# Show the difference between import styles

# Method 1: Clear but verbose
import json as json_module
data1 = json_module.loads('{"name": "Alice", "age": 30}')

# Method 2: Clean but less clear where loads comes from
from json import loads
data2 = loads('{"name": "Bob", "age": 25}')

print(f"Data 1 (using json_module.loads): {data1}")
print(f"Data 2 (using imported loads): {data2}")

print("\n=== AVOIDING NAME CONFLICTS ===")
# Demonstrate potential naming conflicts

# Safe approach - using prefixes
import math
import statistics

# Both modules have functions that could conflict
math_mean = statistics.mean([1, 2, 3, 4, 5])
math_value = math.sqrt(16)

print(f"Statistical mean: {math_mean}")
print(f"Square root: {math_value}")

# Risky approach (uncomment to see potential issues)
# from math import *
# from statistics import *
# This could cause conflicts if both modules have functions with the same name!
````

### Import Best Practices and Guidelines

**Choose the Right Import Pattern:**

| Pattern | Best For | Example |
|---------|----------|---------|
| `import module` | Many functions, avoiding conflicts | `import math` |
| `import module as alias` | Standard conventions, long names | `import pandas as pd` |
| `from module import func` | Few specific functions | `from math import sqrt` |
| `from package import submodule` | Specialized submodules | `from collections import Counter` |
| `from module import *` | **Avoid** (except interactive testing) | ❌ Don't use |

**Standard Conventions (always follow these):**
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import scipy.stats as stats
```

**General Guidelines:**
1. **Put all imports at the top** of your notebook/script
2. **Group imports**: standard library, third-party, local modules
3. **Use conventional aliases** - everyone expects `pd` for pandas
4. **Avoid `import *`** except for quick interactive testing
5. **Be consistent** within a project
6. **Import only what you need** for better performance and clarity

**Error Handling:**
```python
try:
    import optional_library
except ImportError:
    print("Optional library not available")
    optional_library = None
```
