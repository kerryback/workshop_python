### Introduction to Pandas

**Pandas** is the most important library for data analysis in Python. It provides powerful, flexible data structures that make working with structured data intuitive and efficient.

**Key Features:**
- **DataFrame**: 2D labeled data structure (like a spreadsheet)
- **Series**: 1D labeled array (like a column in a spreadsheet)
- **Data Import/Export**: Read from CSV, Excel, and many other formats
- **Data Cleaning**: Handle missing data, duplicates, and data type conversions
- **Data Manipulation**: Filter, sort, group, merge, and transform data
- **Statistical Analysis**: Built-in statistical functions and operations
- **Visualization**: Built-in plotting methods

### Creating DataFrames

There are several ways to create pandas DataFrames depending on your data source and format.

**Common Methods:**
1. **From Dictionary**: `pd.DataFrame(dict)`
2. **From List of Lists**: `pd.DataFrame(data, columns=column_names)`
3. **From CSV File**: `pd.read_csv('file.csv')`
4. **From Excel File**: `pd.read_excel('file.xlsx')`
5. **Empty DataFrame**: `pd.DataFrame()`

**Key Parameters:**
- `columns`: Specify column names
- `index`: Specify row labels
- `dtype`: Specify data types for columns

A common set-up is that columns represent variables and rows represent different observations of the variables.

You don't have to specify index and column names.  If you do not, they will be named 0, 1, 2, ....  It is fairly common to use these default values for the index, but columns are usually named to identify the variables.

````python
# Run this cell to see different ways of creating DataFrames

print("=== Creating DataFrames ===\n")

# Method 1: From dictionary (most common)
dict_data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard'],
    'Price': [999.99, 29.99, 79.99],
    'Stock': [10, 50, 25]
}
df1 = pd.DataFrame(dict_data)
print("1. From Dictionary:")
print(df1)

# Method 2: From list of lists
list_data = [
    ['Apple', 'Fruit', 1.50],
    ['Carrot', 'Vegetable', 0.75],
    ['Bread', 'Bakery', 2.25]
]
df2 = pd.DataFrame(list_data, columns=['Item', 'Category', 'Price'])
print("\n2. From List of Lists:")
print(df2)

# Method 3: From list of dictionaries
dict_list = [
    {'Name': 'John', 'Score': 85, 'Grade': 'B'},
    {'Name': 'Jane', 'Score': 92, 'Grade': 'A'},
    {'Name': 'Bob', 'Score': 78, 'Grade': 'C'}
]
df3 = pd.DataFrame(dict_list)
print("\n3. From List of Dictionaries:")
print(df3)

# Method 4: With custom index
df4 = pd.DataFrame(
    {'Temperature': [20, 25, 30, 22], 'Humidity': [60, 65, 70, 55]},
    index=['Monday', 'Tuesday', 'Wednesday', 'Thursday']
)
print("\n4. With Custom Index:")
print(df4)

# Method 5: Empty DataFrame (useful for building data iteratively)
df5 = pd.DataFrame(columns=['A', 'B', 'C'])
print("\n5. Empty DataFrame:")
print(df5)
````

### Understanding Series - The Building Block of DataFrames

Before diving deeper into DataFrames, it's important to understand **Series** - the fundamental data structure that makes up DataFrame columns.

**What is a Series?**
- A one-dimensional labeled array
- Can contain any data type (integers, strings, floats, objects)
- Each Series has an index that labels each element
- Think of it as a single column of a spreadsheet with row labels

**Key Characteristics:**
- **Homogeneous**: All elements have the same data type
- **Indexed**: Each value has a corresponding label (index)
- **Size Immutable**: Length cannot be changed after creation
- **Data Mutable**: Values can be changed

**Series vs DataFrame:**
- **Series**: 1D (one column)
- **DataFrame**: 2D (multiple columns, each column is a Series)

````python
# Run this cell to explore Series

print("=== Creating and Working with Series ===\n")

# Create a Series from a list
prices = pd.Series([29.99, 999.99, 79.99, 299.99])
print("1. Series from list (with default index):")
print(prices)
print(f"Type: {type(prices)}")

# Create a Series with custom index
products = pd.Series([29.99, 999.99, 79.99, 299.99], 
                    index=['Mouse', 'Laptop', 'Keyboard', 'Monitor'])
print("\n2. Series with custom index:")
print(products)

# Access Series properties
print(f"\n3. Series properties:")
print(f"Values: {products.values}")
print(f"Index: {products.index.tolist()}")
print(f"Data type: {products.dtype}")
print(f"Shape: {products.shape}")
print(f"Size: {products.size}")

# Series from DataFrame column
print(f"\n4. Extracting Series from DataFrame:")
# Using the df from earlier
name_series = df['Name']
print(f"Names column as Series:")
print(name_series)
print(f"Type: {type(name_series)}")
````

### Accessing Series Elements and Ranges

Just like lists, **Series allow you to extract individual elements or ranges of elements** using indexing and slicing. However, Series are more flexible because they support both **label-based indexing** and **position-based indexing**.

**Two Ways to Access Series Data:**
1. **Label-based access** - using the custom index labels you've assigned
2. **Position-based access** - using integer positions (like lists)

**Label-based indexing:**
- **Single element**: `series[label]`
- **Multiple elements**: `series[[label1, label2]]` 
- **Slicing**: `series[start_label:end_label]` (inclusive of both ends!)

**Position-based indexing with `.iloc`:**
- **Single element**: `series.iloc[position]`
- **Multiple elements**: `series.iloc[[pos1, pos2]]`
- **Slicing**: `series.iloc[start:end]` (excludes end, just like lists)
- **Advanced slicing**: `series.iloc[start:end:step]`

**Key Difference from Lists:**
- **Lists**: Only integer positions (`my_list[0]`, `my_list[1:3]`)
- **Series**: Both labels AND positions (`series['label']` OR `series.iloc[0]`)

The `.iloc` accessor makes Series behave exactly like lists for position-based access!

````python
# Run this cell to see Series indexing and slicing in action

# Create a Series with custom index for demonstration
prices = pd.Series([29.99, 999.99, 79.99, 299.99, 149.99, 399.99], 
                   index=['Mouse', 'Laptop', 'Keyboard', 'Monitor', 'Headset', 'Tablet'])
print("=== Series Indexing and Slicing ===\n")
print("Sample Series:")
print(prices)

print("\n=== LABEL-BASED INDEXING ===")
# Single element by label
print(f"1. Single element by label:")
print(f"   prices['Laptop'] = ${prices['Laptop']}")

# Multiple elements by labels
print(f"\n2. Multiple elements by labels:")
selected_items = prices[['Mouse', 'Keyboard', 'Monitor']]
print(selected_items)

# Slicing by labels (INCLUSIVE of both ends)
print(f"\n3. Slicing by labels (Mouse to Monitor - INCLUSIVE):")
price_slice = prices['Mouse':'Monitor']  # This includes both Mouse AND Monitor
print(price_slice)

print("\n=== POSITION-BASED INDEXING WITH .iloc ===")
# Single element by position (like lists!)
print(f"4. Single element by position:")
print(f"   prices.iloc[0] = ${prices.iloc[0]} (first item)")
print(f"   prices.iloc[-1] = ${prices.iloc[-1]} (last item)")

# Multiple elements by positions
print(f"\n5. Multiple elements by positions:")
selected_positions = prices.iloc[[0, 2, 4]]  # positions 0, 2, 4
print(selected_positions)

# Slicing by positions (EXCLUDES end, like lists)
print(f"\n6. Slicing by positions (0:3 - EXCLUDES position 3):")
position_slice = prices.iloc[0:3]  # positions 0, 1, 2 (excludes 3)
print(position_slice)


print("\n=== COMPARISON: SERIES vs LISTS ===")
# Convert to list to show the difference
price_list = prices.tolist()
print(f"9. As a list (position-only access):")
print(f"   price_list[0] = ${price_list[0]}")
print(f"   price_list[1:3] = {price_list[1:3]}")
# print(f"   price_list['Laptop']")  # This would cause an error!

print(f"\n10. Series offers BOTH ways:")
print(f"    prices['Laptop'] = ${prices['Laptop']} (by label)")
print(f"    prices.iloc[1] = ${prices.iloc[1]} (by position - same result!)")

print("\n=== PRACTICAL TIP ===")
print("Use .iloc when you want to treat Series like lists:")
print("- First 3 items: prices.iloc[:3]")
print("- Last 2 items: prices.iloc[-2:]")
print("- Every other item: prices.iloc[::2]")
print("\nUse label indexing when you know the specific names:")
print("- Specific products: prices[['Laptop', 'Mouse']]")
print("- Range by name: prices['Keyboard':'Monitor']")
````

### Setting and Resetting Indexes

The **index** is a crucial component of pandas DataFrames that acts as row labels, similar to row numbers in Excel but much more powerful. Understanding how to work with indexes is essential for effective data manipulation.

**What is an Index?**
- **Row identifier**: Each row has a unique label (like a primary key in databases)
- **Default**: When you create a DataFrame, pandas assigns integer index 0, 1, 2, ...
- **Custom**: You can use meaningful labels like dates, names, or IDs
- **Alignment**: Indexes enable automatic data alignment in operations

**Why Change the Index?**
- **Meaningful labels**: Use dates, product IDs, or names instead of numbers
- **Better data organization**: Group related data together
- **Easier data access**: Access rows by meaningful names
- **Time series analysis**: Date indexes unlock powerful time-based operations
- **Performance**: Indexes can speed up data lookup and operations

**Key Index Operations:**
- **Set index**: `df.set_index('column_name')`
- **Reset index**: `df.reset_index()`
- **Multiple indexes**: `df.set_index(['col1', 'col2'])`
- **Drop current index**: `df.reset_index(drop=True)`

````python
# Run this cell to explore index operations

# Create a sample business dataset with default index
business_data = pd.DataFrame({
    'Company': ['Apple', 'Microsoft', 'Google', 'Amazon', 'Tesla', 'Meta'],
    'Industry': ['Technology', 'Technology', 'Technology', 'E-commerce', 'Automotive', 'Technology'],
    'Revenue_Billions': [394.3, 198.3, 282.8, 513.9, 81.5, 117.9],
    'Employees': [164000, 221000, 156000, 1608000, 127855, 86482],
    'Founded': [1976, 1975, 1998, 1994, 2003, 2004]
})

print("=== Original DataFrame (Default Integer Index) ===")
print(business_data)
print(f"Current index: {business_data.index.tolist()}")
print(f"Index name: {business_data.index.name}")

print("\n=== Setting Company as Index ===")
# Set 'Company' column as the index
company_indexed = business_data.set_index('Company')
print(company_indexed)
print(f"New index: {company_indexed.index.tolist()}")
print(f"Index name: {company_indexed.index.name}")

# Now we can access rows by company name!
print(f"\n1. Access Apple's data using company name:")
print(company_indexed.loc['Apple'])

print(f"\n2. Access multiple companies:")
print(company_indexed.loc[['Apple', 'Tesla']])

print(f"\n3. Revenue for Google:")
print(f"${company_indexed.loc['Google', 'Revenue_Billions']} billion")

print("\n=== Resetting Index (Back to Default) ===")
# Reset index - 'Company' becomes a regular column again
reset_df = company_indexed.reset_index()
print(reset_df)
print(f"Index after reset: {reset_df.index.tolist()}")

print("\n=== Setting Index with Drop Option ===")
# Sometimes you want to set index but keep the original column
company_indexed_keep = business_data.set_index('Company', drop=False)
print("Set index but kept 'Company' as column too:")
print(company_indexed_keep.head(3))

print("\n=== Reset Index with Drop Option ===")
# When resetting, you can choose to drop the current index
dropped_reset = company_indexed.reset_index(drop=True)
print("Reset index and dropped the previous index:")
print(dropped_reset)

print("\n=== Multiple Index (Hierarchical) ===")
# Create a dataset with multiple index levels
sales_data = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South', 'North', 'South'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Phone', 'Tablet', 'Tablet'],
    'Quarter': ['Q1', 'Q1', 'Q1', 'Q1', 'Q2', 'Q2'],
    'Sales': [1500, 2000, 1200, 1800, 1000, 900]
})

# Set multiple columns as index
multi_indexed = sales_data.set_index(['Region', 'Product'])
print("Multiple index (Region and Product):")
print(multi_indexed)

print(f"\n4. Access North region data:")
print(multi_indexed.loc['North'])

print(f"\n5. Access specific product in specific region:")
print(multi_indexed.loc[('North', 'Laptop')])

print("\n=== Practical Example: Time Series Index ===")
# Create a time series example
from datetime import datetime, timedelta

dates = pd.date_range(start='2024-01-01', periods=6, freq='M')
stock_data = pd.DataFrame({
    'Date': dates,
    'Stock_Price': [150.25, 155.80, 148.90, 162.45, 159.30, 167.75],
    'Volume': [1000000, 1200000, 950000, 1300000, 1100000, 1250000]
})

print("Before setting date index:")
print(stock_data)

# Set Date as index for time series operations
stock_indexed = stock_data.set_index('Date')
print("\nAfter setting Date as index:")
print(stock_indexed)

print(f"\n6. Access data for specific month:")
print(stock_indexed.loc['2024-03-31'])

print(f"\n7. Slice data by date range:")
print(stock_indexed.loc['2024-02-29':'2024-05-31'])

print("\n=== Key Takeaways ===")
print("• Default index: 0, 1, 2, ... (good for general use)")
print("• Custom index: Meaningful labels (better for specific data access)")
print("• set_index(): Convert column to index")
print("• reset_index(): Convert index back to column")
print("• Multiple indexes: Create hierarchical data structure")
print("• Date indexes: Unlock powerful time series operations")
````

### Selecting and Accessing Data

Pandas provides multiple ways to select and access data from DataFrames, similar to how you might work with spreadsheets or databases.

**Column Selection:**
- **Single column**: `df['column_name']` or `df.column_name`
- **Multiple columns**: `df[['col1', 'col2']]`

**Row Selection:**
- **By index**: `df.loc[row_label]` or `df.iloc[row_number]`
- **By range**: `df.loc[start:end]` or `df.iloc[start:end]`

**Cell Selection:**
- **Specific cell**: `df.loc[row, column]` or `df.iloc[row_num, col_num]`
- **Multiple cells**: `df.loc[rows, columns]`

**Boolean Indexing:**
- **Filter rows**: `df[df['column'] > value]`
- **Multiple conditions**: `df[(df['col1'] > val1) & (df['col2'] == val2)]`

````python
# Run this cell to see data selection methods

# Create a sample DataFrame for examples
employees = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'Age': [25, 30, 35, 28, 32, 45],
    'Department': ['Engineering', 'Marketing', 'Engineering', 'HR', 'Marketing', 'Engineering'],
    'Salary': [70000, 65000, 80000, 60000, 72000, 90000],
    'Years': [2, 5, 8, 3, 6, 12]
})

print("Sample DataFrame:")
print(employees)
print()

print("=== Column Selection ===")
# Single column (returns Series)
print("1. Single column (Name):")
print(employees['Name'])
print(f"Type: {type(employees['Name'])}")

print("\n2. Single column using dot notation:")
print(employees.Department.head(3))

# Multiple columns (returns DataFrame)
print("\n3. Multiple columns:")
print(employees[['Name', 'Salary']])

print("\n=== Row Selection ===")
# Single row by position
print("4. Single row by position (iloc):")
print(employees.iloc[0])  # First row

print("\n5. Multiple rows by position:")
print(employees.iloc[1:4])  # Rows 1, 2, 3

print("\n6. Multiple rows by index labels:")
print(employees.loc[0:2])  # Rows 0, 1, 2 (inclusive)

print("\n=== Cell Selection ===")
print("7. Specific cell:")
print(f"Alice's salary: {employees.loc[0, 'Salary']}")
print(f"Same using iloc: {employees.iloc[0, 3]}")

print("\n8. Multiple cells:")
print(employees.loc[0:2, ['Name', 'Age']])

print("\n=== Boolean Indexing (Filtering) ===")
print("9. Employees with salary > 70000:")
high_earners = employees[employees['Salary'] > 70000]
print(high_earners)

print("\n10. Multiple conditions (Engineering dept with salary > 70000):")
eng_high = employees[(employees['Department'] == 'Engineering') & (employees['Salary'] > 70000)]
print(eng_high)

print("\n11. Using isin() for multiple values:")
marketing_hr = employees[employees['Department'].isin(['Marketing', 'HR'])]
print(marketing_hr)
````

### Data Manipulation and Modification

Pandas provides extensive capabilities for modifying and manipulating data within DataFrames.

**Adding and Modifying Columns:**
- **New column**: `df['new_col'] = values`
- **Calculated column**: `df['new_col'] = df['col1'] + df['col2']`
- **Conditional column**: `df['new_col'] = np.where(condition, value1, value2)`

**Adding and Removing Rows:**
- **Add row**: `df.loc[new_index] = values`
- **Concatenate**: `pd.concat([df1, df2])`
- **Drop rows**: `df.drop(index_labels)`

**Modifying Values:**
- **Single value**: `df.loc[row, col] = new_value`
- **Multiple values**: `df.loc[condition, 'column'] = new_value`
- **Replace values**: `df.replace(old_value, new_value)`

**Column Operations:**
- **Rename columns**: `df.rename(columns={'old': 'new'})`
- **Drop columns**: `df.drop(columns=['col1', 'col2'])`
- **Reorder columns**: `df[['col2', 'col1', 'col3']]`

````python
# Run this cell to see basic data manipulation techniques

# Start with a copy of our employees DataFrame
emp_df = employees.copy()
print("Original DataFrame:")
print(emp_df)

print("\n=== Adding New Columns ===")
# Calculate annual bonus (10% of salary)
emp_df['Bonus'] = emp_df['Salary'] * 0.10
print("1. Added Bonus column:")
print(emp_df[['Name', 'Salary', 'Bonus']])

# Add experience level based on years using pandas cut
emp_df['Experience_Level'] = pd.cut(emp_df['Years'], 
                                   bins=[0, 5, 10, float('inf')], 
                                   labels=['Junior', 'Senior', 'Expert'], 
                                   right=False)

print("\n2. Added Experience Level:")
print(emp_df[['Name', 'Years', 'Experience_Level']])

print("\n=== Modifying Existing Data ===")
# Give everyone a 5% raise
emp_df['Salary'] = emp_df['Salary'] * 1.05
print("3. After 5% salary increase:")
print(emp_df[['Name', 'Salary']])

# Update specific values
emp_df.loc[emp_df['Name'] == 'Alice', 'Department'] = 'Data Science'
print("\n4. Alice moved to Data Science:")
print(emp_df[emp_df['Name'] == 'Alice'])

print("\n=== Column Operations ===")
# Rename columns
emp_df_renamed = emp_df.rename(columns={'Years': 'Years_Experience', 'Bonus': 'Annual_Bonus'})
print("5. Renamed columns:")
print(list(emp_df_renamed.columns))

# Drop columns
emp_df_reduced = emp_df.drop(columns=['Bonus', 'Experience_Level'])
print("\n6. After dropping columns:")
print(list(emp_df_reduced.columns))
````
