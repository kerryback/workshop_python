### Notebooks

- The Colab interface is a **Jupyter notebook**.
- A notebook is a collection of cells. 
- Cells contain either explanatory text or executable code and its output. 
- You can add new cells by using the `+ CODE` and `+ TEXT` buttons in the toolbar or the buttons that show when you hover between cells. 



### Assignment statements 

Type the following in a cell.  Then hit the Play/Run button at the left side of the cell to execute it.

```python
x = 10
print(x)
```

The statement `x=10` assigns the value 10 to the variable `x`.  The statement `print(x)` prints the value of `x` to the screen.

Add a new code cell and type

```python
y = 2 * x
print(y)
```

- Spaces around the `=` and `+` are optional.
- Now we have two variables in our Python session.  They are independent.  If we change `x`, it will not affect `y`.  Try 

```python 
x = 0 
print(y)
```

### Notebooks vs spreadsheets

- Notebooks are similar to spreadsheets in some ways but differ in others.  
- Assigning the value 10 to `x` and defining `y` as `2*x` is similar to having the value 10 in cell `A1` in a spreadsheet and putting `=2*A1` in cell `B1.`  
- The cell names in a spreadsheet are therefore similar to variable names in python.


If it becomes confusing as to what has executed and what hasn't, and you want to get back to the order in the notebook, select a cell, use **Runtime/Restart Session** to clear python, and then use **Runtime/Run Before** to run all cells that appear in the notebook before the selected cell.

- An important difference between notebooks and spreadsheets is that all formulas in a spreadsheet become active when we open it, but nothing is activated when we open a notebook.  

- When we open a notebook, we are just getting a image of a text file rendered by our browser.  

- Python has not yet done anything.  It doesn't know any variables and it doesn't know any formulas.  

- Python becomes active when we execute a code cell.  Only the executed code cells are meaningful.  
- Code cells are meaningful **in the order they are executed, not in the order in which they appear in the notebook.**  
- This is important to keep in mind when moving around in a notebook.

### Self-grading cells

In subsequent parts of this workshop, we'll use self-grading code cells.  They have the following structure:

* a few lines beginning with `#`
* blank lines
* a few lines beginning with `assert`

Anything following a `#` on a line of a code cell is ignored by the python interpreter.  They are comments intended for people to read.  

* The first few lines of a self-grading cell are instructions.
* Follow the instructions and write code at the blank line.  You can add more code lines if needed using the `Enter` key as usual.
* The lines beginning with `assert` check your answer.  You do not need to read those lines.

If you have difficulty with any self-grading cell in this workshop, ask the chatbot, your instructor, or a TA for help.
