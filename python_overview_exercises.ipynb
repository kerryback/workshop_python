## Pattern ①: Objects & Types

Everything in Python is an **object** with a **type**.

```python
# Types of objects
print(type(42))         # int
print(type(3.14))       # float
print(type("hello"))    # str
print(type(True))       # bool
```

```python
# String methods — methods are actions an object can perform
ticker = "aapl"
print(ticker.upper())           # AAPL
print(ticker.replace("a", "X")) # XXpl

messy = "  Apple Inc.  "
print(messy.strip())            # remove whitespace
```

```python
# Booleans & comparisons
revenue = 500_000
target  = 400_000

print(revenue > target)   # True
print(revenue == target)  # False
print(revenue >= target)  # True
```

## Pattern ②: Variables

Variables are **labels** attached to objects. Use `=` to assign a value.

```python
# Variables as labels
price   = 182.50
company = "Apple"
shares  = 1_000

market_cap = price * shares
print(f"{company} market cap: ${market_cap:,.2f}")
```

```python
# Math operations
revenue = 1_000_000
costs   = 750_000

profit = revenue - costs
margin = profit / revenue
growth = 1.08 ** 5      # 8% compounded for 5 years

print(f"Profit: ${profit:,}")
print(f"Margin: {margin:.0%}")
print(f"Growth factor: {growth:.4f}")
```

## Pattern ③: Collections

**Lists** and **dictionaries** hold multiple pieces of data.

```python
# Lists — ordered, indexed from 0
tickers = ["AAPL", "GOOGL", "MSFT", "AMZN"]

print(tickers[0])    # first item
print(tickers[-1])   # last item
print(tickers[1:3])  # slice
print(len(tickers))  # length
```

```python
# Dictionaries — key/value pairs
stock = {
    "ticker": "AAPL",
    "price":  182.50,
    "sector": "Technology"
}

print(stock["ticker"])
print(stock["price"])
```

## Pattern ④: Conditionals

Programs make **decisions** with `if`, `elif`, `else`.

```python
# if / elif / else
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
```

```python
# Logical operators: and, or, not
age    = 30
income = 120_000

if age >= 25 and income >= 100_000:
    print("Eligible for premium card")

# Membership testing with 'in'
tech_stocks = ["AAPL", "GOOGL", "MSFT"]
if "AAPL" in tech_stocks:
    print("Apple is in the list")
```

## Pattern ⑤: Loops

`for` loops repeat an action for every item in a collection.

```python
# for loop over a list
tickers = ["AAPL", "GOOGL", "MSFT"]

for ticker in tickers:
    print(f"Looking up {ticker}...")
```

```python
# range() for numeric loops
for i in range(1, 6):
    print(f"{i} squared is {i**2}")
```

```python
# Accumulator pattern
prices = [182.50, 141.80, 378.90, 178.25]

total = 0
for price in prices:
    total += price

average = total / len(prices)
print(f"Average price: ${average:,.2f}")
```

```python
# List comprehension — a compact loop on one line
prices = [182.50, 141.80, 378.90, 178.25]

doubled   = [p * 2 for p in prices]
expensive = [p for p in prices if p > 150]

print(doubled)
print(expensive)
```

## Pattern ⑥: Functions & Methods

**Functions** are reusable blocks of logic. **Methods** are functions that belong to an object.

```python
# Built-in functions
numbers = [42, 17, 85, 23, 61]

print("Length:", len(numbers))
print("Max:",    max(numbers))
print("Min:",    min(numbers))
print("Sum:",    sum(numbers))
print("Sorted:", sorted(numbers))
```

```python
# Defining your own function
def calculate_roi(cost, revenue):
    profit = revenue - cost
    roi    = profit / cost
    return roi

result = calculate_roi(100_000, 150_000)
print(f"ROI: {result:.0%}")
```

```python
# Default arguments
def format_currency(amount, currency="USD", decimals=2):
    if currency == "USD":
        return f"${amount:,.{decimals}f}"
    else:
        return f"{amount:,.{decimals}f} {currency}"

print(format_currency(1_234_567))
print(format_currency(1_234_567, currency="EUR"))
print(format_currency(1_234_567, decimals=0))
```

## Bonus: Libraries & pandas

Libraries extend Python with pre-built tools. `import pandas as pd` loads pandas and gives it the short name `pd`.

```python
import pandas as pd

data = {
    "Ticker": ["AAPL", "GOOGL", "MSFT", "AMZN"],
    "Price":  [182.50, 141.80, 378.90, 178.25],
    "Sector": ["Tech", "Tech", "Tech", "Retail"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Method chaining — each method returns an object
print("Mean price:", df["Price"].mean())
print("Max price:",  df["Price"].max())
print()
print(df.sort_values("Price", ascending=False))
```
