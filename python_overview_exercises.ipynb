## Part 1: Objects

Everything in Python is an **object** with a type.

```python
# Types of objects
print(type(42))         # int
print(type(3.14))       # float
print(type("hello"))    # str
print(type(True))       # bool
```

```python
# Variables as labels
price = 182.50
company = "Apple"
shares = 1_000

market_cap = price * shares
print(f"{company} market cap: ${market_cap:,.2f}")
```

```python
# Math operations
revenue = 1_000_000
costs = 750_000

profit = revenue - costs
margin = profit / revenue
growth = 1.08 ** 5      # 8% compounded for 5 years

print(f"Profit: ${profit:,}")
print(f"Margin: {margin:.0%}")
print(f"Growth factor: {growth:.4f}")
```

```python
# String methods
ticker = "aapl"
print(ticker.upper())
print(ticker.replace("a", "X"))

messy = "  Apple Inc.  "
print(messy.strip())
```

```python
# Booleans & comparisons
revenue = 500_000
target = 400_000

print(revenue > target)     # True
print(revenue == target)    # False
print(revenue >= target)    # True
```

## Part 3: Conditional Execution

Programs make **decisions** with `if`, `elif`, `else`.

```python
# if / elif / else
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
```

```python
# Logical operators: and, or, not
age = 30
income = 120_000

if age >= 25 and income >= 100_000:
    print("Eligible for premium card")

# Membership testing
tech_stocks = ["AAPL", "GOOGL", "MSFT"]
if "AAPL" in tech_stocks:
    print("Apple is in the list")
```

## Part 4: Loops

Loops repeat actions — `for` loops iterate over collections.

```python
# for loop over a list
tickers = ["AAPL", "GOOGL", "MSFT"]

for ticker in tickers:
    print(f"Looking up {ticker}...")
```

```python
# range() for numeric loops
for i in range(1, 6):
    print(f"{i} squared is {i**2}")
```

```python
# Accumulator pattern
prices = [182.50, 141.80, 378.90, 178.25]

total = 0
for price in prices:
    total += price

average = total / len(prices)
print(f"Average price: ${average:,.2f}")
```

```python
# List comprehension
prices = [182.50, 141.80, 378.90, 178.25]

doubled = [p * 2 for p in prices]
print(doubled)

# With filtering
expensive = [p for p in prices if p > 150]
print(expensive)
```

## Part 5: Functions

Functions are reusable blocks of logic.

```python
# Built-in functions
numbers = [42, 17, 85, 23, 61]

print("Length:", len(numbers))
print("Max:", max(numbers))
print("Min:", min(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))
```

```python
# Defining a function
def calculate_roi(cost, revenue):
    profit = revenue - cost
    roi = profit / cost
    return roi

result = calculate_roi(100_000, 150_000)
print(f"ROI: {result:.0%}")
```

```python
# Default arguments
def format_currency(amount, currency="USD", decimals=2):
    if currency == "USD":
        return f"${amount:,.{decimals}f}"
    else:
        return f"{amount:,.{decimals}f} {currency}"

print(format_currency(1_234_567))
print(format_currency(1_234_567, currency="EUR"))
print(format_currency(1_234_567, decimals=0))
```

## Part 6: Libraries & pandas

Libraries extend Python with pre-built tools.

```python
import pandas as pd

data = {
    "Ticker": ["AAPL", "GOOGL", "MSFT", "AMZN"],
    "Price": [182.50, 141.80, 378.90, 178.25],
    "Sector": ["Tech", "Tech", "Tech", "Retail"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Basic DataFrame operations
print("Mean price:", df["Price"].mean())
print("Max price:", df["Price"].max())
print()
print(df.sort_values("Price", ascending=False))
```
